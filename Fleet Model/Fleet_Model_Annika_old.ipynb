```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

basisjahr = 2025
zieljahr = 2050

# Prozentuale Veränderung gegenüber dem Basisbestand
# Beispiele:
# +10 %  ->  10
# -20 %  -> -20
veraenderung_prozent = 100

# Art der Fortschreibung:
# "linear"  = jedes Jahr gleicher absoluter Zuwachs/Rückgang
# "growth"  = jedes Jahr gleiche prozentuale Wachstumsrate
fortschreibung = "growth"


# ------------------------------------------------------------
# 2. Gesamtbestand je Jahr berechnen
# ------------------------------------------------------------

bestand_total_by_year = (
    bestand_df
    .groupby("Berichtsjahr", as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Berichtsjahr": "Jahr", "Anzahl": "gesamtbestand"})
)


# ------------------------------------------------------------
# 3. Basisbestand bestimmen
# ------------------------------------------------------------

basisbestand = bestand_total_by_year.loc[
    bestand_total_by_year["Jahr"] == basisjahr,
    "gesamtbestand"
].iloc[0]


# ------------------------------------------------------------
# 4. Zielbestand berechnen
# ------------------------------------------------------------

zielbestand = basisbestand * (1 + veraenderung_prozent / 100)


# ------------------------------------------------------------
# 5. Zeitreihe vom Basisjahr bis Zieljahr erzeugen
# ------------------------------------------------------------

jahre = np.arange(basisjahr, zieljahr + 1)
anzahl_jahre = zieljahr - basisjahr

if fortschreibung == "linear":
    # Jedes Jahr gleicher absoluter Zuwachs/Rückgang
    gesamtbestand_target = np.linspace(
        basisbestand,
        zielbestand,
        len(jahre)
    )

elif fortschreibung == "growth":
    # Jedes Jahr gleiche prozentuale Wachstumsrate
    wachstumsrate = (zielbestand / basisbestand) ** (1 / anzahl_jahre) - 1

    gesamtbestand_target = basisbestand * (
        (1 + wachstumsrate) ** np.arange(len(jahre))
    )

else:
    raise ValueError("fortschreibung muss entweder 'linear' oder 'growth' sein.")


# ------------------------------------------------------------
# 6. Output-DataFrame erstellen
# ------------------------------------------------------------

fleet_total_df = pd.DataFrame({
    "Jahr": jahre,
    "gesamtbestand_target": gesamtbestand_target
})


# ------------------------------------------------------------
# 7. Kontrolle ausgeben
# ------------------------------------------------------------

print("Basisjahr:", basisjahr)
print("Basisbestand:", basisbestand)

print("Zieljahr:", zieljahr)
print("Zielbestand:", zielbestand)

print("Fortschreibung:", fortschreibung)

print("\nErste Zeilen:")
display(fleet_total_df.head())

print("\nLetzte Zeilen:")
display(fleet_total_df.tail())
```

```python
import matplotlib.pyplot as plt

# Liniendiagramm für die Entwicklung des Gesamtbestands
ax = fleet_total_df.plot(
    x="Jahr",
    y="gesamtbestand_target",
    kind="line",
    marker="o",
    figsize=(12, 6),
    legend=False
)

ax.set_xlabel("Year")
ax.set_ylabel("Total stock")
ax.set_title("Development of total vehicle stock")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

anzahl_betrachtete_jahre = 5
altersklassen_breite = 5

gruppen_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung"
]


# ------------------------------------------------------------
# 1. Arbeitskopien erstellen
# ------------------------------------------------------------

abs_work = abs_df.copy()
bestand_work = bestand_df.copy()


# ------------------------------------------------------------
# 2. Relevante Spalten prüfen und bereinigen
# ------------------------------------------------------------

relevante_spalten = gruppen_spalten + ["Anzahl"]

abs_work = abs_work[relevante_spalten].copy()
bestand_work = bestand_work[relevante_spalten].copy()

# Zahlen-/Jahresspalten sicher numerisch machen
for df in [abs_work, bestand_work]:
    df["Berichtsjahr"] = pd.to_numeric(df["Berichtsjahr"], errors="coerce")
    df["Jahr der Erstzulassung"] = pd.to_numeric(
        df["Jahr der Erstzulassung"],
        errors="coerce"
    )
    df["Anzahl"] = pd.to_numeric(df["Anzahl"], errors="coerce")

# Fehlende oder ungültige Werte ausschließen
abs_work = abs_work.dropna(subset=relevante_spalten)
bestand_work = bestand_work.dropna(subset=relevante_spalten)

# Jahre als Integer speichern
for df in [abs_work, bestand_work]:
    df["Berichtsjahr"] = df["Berichtsjahr"].astype(int)
    df["Jahr der Erstzulassung"] = df["Jahr der Erstzulassung"].astype(int)


# ------------------------------------------------------------
# 3. Letzte verfügbare Jahre der Abgangsdaten bestimmen
# ------------------------------------------------------------

aktuelles_jahr = abs_work["Berichtsjahr"].max()
startjahr = aktuelles_jahr - anzahl_betrachtete_jahre + 1

basisjahre = list(range(startjahr, aktuelles_jahr + 1))

print(f"Basisjahre: {startjahr}-{aktuelles_jahr}")


# ------------------------------------------------------------
# 4. Nur betrachtete Jahre verwenden
# ------------------------------------------------------------

abs_work = abs_work[abs_work["Berichtsjahr"].isin(basisjahre)].copy()
bestand_work = bestand_work[bestand_work["Berichtsjahr"].isin(basisjahre)].copy()


# ------------------------------------------------------------
# 5. Abgänge und Bestände auf gemeinsame Gruppierung aggregieren
# ------------------------------------------------------------

abs_grouped = (
    abs_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "abgaenge"})
)

bestand_grouped = (
    bestand_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand"})
)


# ------------------------------------------------------------
# 6. Abgänge mit passendem Bestand verbinden
# ------------------------------------------------------------

hazard_base = abs_grouped.merge(
    bestand_grouped,
    on=gruppen_spalten,
    how="inner"
)


# ------------------------------------------------------------
# 7. Fahrzeugalter berechnen und ungültige Fälle ausschließen
# ------------------------------------------------------------

hazard_base["Alter"] = (
    hazard_base["Berichtsjahr"] - hazard_base["Jahr der Erstzulassung"]
)

hazard_base = hazard_base[
    (hazard_base["bestand"].notna()) &
    (hazard_base["bestand"] > 0) &
    (hazard_base["Alter"] >= 0)
].copy()


# ------------------------------------------------------------
# 8. Fahrzeugalter in 5-Jahres-Altersklassen einteilen
# ------------------------------------------------------------

hazard_base["Altersklasse_Start"] = (
    hazard_base["Alter"] // altersklassen_breite
) * altersklassen_breite

hazard_base["Altersklasse_Ende"] = (
    hazard_base["Altersklasse_Start"] + altersklassen_breite - 1
)

hazard_base["Altersklasse"] = (
    hazard_base["Altersklasse_Start"].astype(str)
    + "-"
    + hazard_base["Altersklasse_Ende"].astype(str)
)


# ------------------------------------------------------------
# 9. Gepoolte Hazard-Rate berechnen
# ------------------------------------------------------------

hazard_pooled_df = (
    hazard_base
    .groupby(
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "Altersklasse"
        ],
        as_index=False
    )
    .agg(
        summe_abgaenge=("abgaenge", "sum"),
        summe_bestand=("bestand", "sum"),
        anzahl_betrachtete_jahre=("Berichtsjahr", "nunique")
    )
)

hazard_pooled_df["pooled_hazard_rate"] = (
    hazard_pooled_df["summe_abgaenge"] / hazard_pooled_df["summe_bestand"]
)


# ------------------------------------------------------------
# 10. Optional: Sortierung
# ------------------------------------------------------------

hazard_pooled_df = hazard_pooled_df.sort_values(
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start"
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 11. Kontrolle anzeigen
# ------------------------------------------------------------

print(f"Basisjahre: {startjahr}-{aktuelles_jahr}")
print(f"Anzahl betrachtete Jahre: {anzahl_betrachtete_jahre}")
print(f"Breite der Altersklassen: {altersklassen_breite} Jahre")

display(hazard_pooled_df.head())
display(hazard_pooled_df.tail())
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

anzahl_betrachtete_jahre = 5
altersklassen_breite = 5
max_reentry_quote = 0.9

# Nur Altersklassen bis 20-24 berücksichtigen
max_alter = 24

gruppen_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung"
]


# ------------------------------------------------------------
# Hilfsfunktion: Berichtsjahr bereinigen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr"):
    """
    Wandelt Berichtsjahr robust in ein numerisches Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Falls ein vollständiges Datum vorliegt, wird das Kalenderjahr extrahiert.
    Hinweis:
    Wenn dein Bestand mit 01.01.2026 das Basisjahr 2025 beschreibt,
    wird für bestand_df unten explizit minus 1 gerechnet.
    """
    s = df[spalte]

    # Erst als Zahl versuchen
    jahr_num = pd.to_numeric(s, errors="coerce")

    # Für nicht-numerische Werte als Datum versuchen
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    return jahr_num.fillna(jahr_date).astype("Int64")


# ------------------------------------------------------------
# 1. Arbeitskopien erstellen
# ------------------------------------------------------------

abs_work = abs_df.copy()
nzl_work = nzl_df.copy()
bestand_work = bestand_df.copy()


# ------------------------------------------------------------
# 2. Exits aus abs_df vorbereiten und aggregieren
# ------------------------------------------------------------

abs_relevante_spalten = gruppen_spalten + ["Anzahl"]

abs_work = abs_work[abs_relevante_spalten].copy()

abs_work["Berichtsjahr"] = bereinige_berichtsjahr(abs_work, "Berichtsjahr")
abs_work["Jahr der Erstzulassung"] = pd.to_numeric(
    abs_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")
abs_work["Anzahl"] = pd.to_numeric(abs_work["Anzahl"], errors="coerce")

abs_work = abs_work.dropna(subset=abs_relevante_spalten)

exits_grouped = (
    abs_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "exits"})
)


# ------------------------------------------------------------
# 3. Neuzulassungen aus nzl_df vorbereiten und aggregieren
# ------------------------------------------------------------

nzl_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Anzahl"
]

nzl_work = nzl_work[nzl_relevante_spalten].copy()

nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(nzl_work, "Berichtsjahr")
nzl_work["Anzahl"] = pd.to_numeric(nzl_work["Anzahl"], errors="coerce")

# Neuzulassungen im Jahr t werden als Kohorte Erstzulassungsjahr t interpretiert
nzl_work["Jahr der Erstzulassung"] = nzl_work["Berichtsjahr"]

nzl_work = nzl_work.dropna(subset=gruppen_spalten + ["Anzahl"])

entries_grouped = (
    nzl_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "entries"})
)


# ------------------------------------------------------------
# 4. Bestände aus bestand_df vorbereiten und aggregieren
# ------------------------------------------------------------

bestand_relevante_spalten = gruppen_spalten + ["Anzahl"]

bestand_work = bestand_work[bestand_relevante_spalten].copy()

bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
    bestand_work,
    "Berichtsjahr"
)

# WICHTIG:
# Falls bestand_df in "Berichtsjahr" Datumswerte wie 01.01.2026 enthält,
# beschreibt dieser Bestand das Basisjahr 2025.
# Deshalb wird beim Bestand minus 1 gerechnet.
bestand_work["Berichtsjahr"] = bestand_work["Berichtsjahr"] - 1

bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
    bestand_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")
bestand_work["Anzahl"] = pd.to_numeric(bestand_work["Anzahl"], errors="coerce")

bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

stock_grouped = (
    bestand_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand"})
)


# ------------------------------------------------------------
# 5. Bestand t und Bestand t+1 derselben Kohorte zusammenführen
# ------------------------------------------------------------

stock_t = stock_grouped.rename(columns={"bestand": "bestand_t"}).copy()

stock_t_plus_1 = stock_grouped.copy()
stock_t_plus_1["Berichtsjahr"] = stock_t_plus_1["Berichtsjahr"] - 1
stock_t_plus_1 = stock_t_plus_1.rename(columns={"bestand": "bestand_t_plus_1"})

balance_df = stock_t.merge(
    stock_t_plus_1,
    on=gruppen_spalten,
    how="inner"
)


# ------------------------------------------------------------
# 6. Exits und Entries im Jahr t ergänzen
# ------------------------------------------------------------

balance_df = balance_df.merge(
    exits_grouped,
    on=gruppen_spalten,
    how="left"
)

balance_df = balance_df.merge(
    entries_grouped,
    on=gruppen_spalten,
    how="left"
)

balance_df["exits"] = balance_df["exits"].fillna(0)
balance_df["entries"] = balance_df["entries"].fillna(0)


# ------------------------------------------------------------
# 7. Reentries aus Bestandsbilanz berechnen
# ------------------------------------------------------------

balance_df["reentries_raw"] = (
    balance_df["bestand_t_plus_1"]
    - balance_df["bestand_t"]
    + balance_df["exits"]
    - balance_df["entries"]
)

balance_df["reentry_raw_negativ"] = balance_df["reentries_raw"] < 0

# Für Modellierung negative Reentries auf 0 setzen
balance_df["reentries"] = balance_df["reentries_raw"].clip(lower=0)


# ------------------------------------------------------------
# 8. Fahrzeugalter berechnen und Altersklassen bilden
# ------------------------------------------------------------

balance_df["Alter"] = (
    balance_df["Berichtsjahr"] - balance_df["Jahr der Erstzulassung"]
)

balance_df = balance_df[
    (balance_df["Alter"] >= 0) &
    (balance_df["Alter"] <= max_alter)
].copy()

balance_df["Altersklasse_Start"] = (
    balance_df["Alter"] // altersklassen_breite
) * altersklassen_breite

balance_df["Altersklasse_Ende"] = (
    balance_df["Altersklasse_Start"] + altersklassen_breite - 1
)

balance_df["Altersklasse"] = (
    balance_df["Altersklasse_Start"].astype(str)
    + "-"
    + balance_df["Altersklasse_Ende"].astype(str)
)


# ------------------------------------------------------------
# 9. Erster Output:
#    Reentries nach Jahr, Segment, Antriebsart und Altersklasse
# ------------------------------------------------------------

reentry_yearly_df = (
    balance_df
    .groupby(
        [
            "Berichtsjahr",
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "Altersklasse"
        ],
        as_index=False
    )
    .agg(
        summe_exits=("exits", "sum"),
        summe_reentries=("reentries", "sum"),
        summe_reentries_raw=("reentries_raw", "sum"),
        anzahl_negative_reentry_raw=("reentry_raw_negativ", "sum")
    )
)

reentry_yearly_df["reentry_quote_exits"] = np.where(
    reentry_yearly_df["summe_exits"] > 0,
    reentry_yearly_df["summe_reentries"] / reentry_yearly_df["summe_exits"],
    0
)

reentry_yearly_df["reentry_quote_exits"] = (
    reentry_yearly_df["reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_yearly_df = reentry_yearly_df.sort_values(
    [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Altersklasse_Start"
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 10. Kalibrierungsfenster bestimmen
# ------------------------------------------------------------

aktuelles_jahr = reentry_yearly_df["Berichtsjahr"].max()
startjahr = aktuelles_jahr - anzahl_betrachtete_jahre + 1

kalibrierungsjahre = list(range(startjahr, aktuelles_jahr + 1))

print(f"Kalibrierungsfenster: {startjahr}-{aktuelles_jahr}")


# ------------------------------------------------------------
# 11. Zweiter Output:
#     gepoolte Reentry-Quote über die letzten Jahre
# ------------------------------------------------------------

reentry_calibration_base = reentry_yearly_df[
    reentry_yearly_df["Berichtsjahr"].isin(kalibrierungsjahre)
].copy()

reentry_pooled_df = (
    reentry_calibration_base
    .groupby(
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "Altersklasse"
        ],
        as_index=False
    )
    .agg(
        summe_reentries=("summe_reentries", "sum"),
        summe_exits=("summe_exits", "sum"),
        anzahl_betrachtete_jahre=("Berichtsjahr", "nunique")
    )
)

reentry_pooled_df["pooled_reentry_quote_exits"] = np.where(
    reentry_pooled_df["summe_exits"] > 0,
    reentry_pooled_df["summe_reentries"] / reentry_pooled_df["summe_exits"],
    0
)

reentry_pooled_df["pooled_reentry_quote_exits"] = (
    reentry_pooled_df["pooled_reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_pooled_df = reentry_pooled_df.sort_values(
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start"
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 12. Kontrolle anzeigen
# ------------------------------------------------------------

print(f"Kalibrierungsfenster: {startjahr}-{aktuelles_jahr}")
print(f"Anzahl betrachtete Jahre: {anzahl_betrachtete_jahre}")
print(f"Breite der Altersklassen: {altersklassen_breite} Jahre")
print(f"Maximale Reentry-Quote: {max_reentry_quote}")

print("\nErster Output: reentry_yearly_df")
display(reentry_yearly_df.head())

print("\nZweiter Output: reentry_pooled_df")
display(reentry_pooled_df.head())
```

# Bestandsfortschreibung von 2025 auf 2026

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

basisjahr = 2025
prognosejahr = 2026

altersklassen_breite = 5

# Maximale Wahrscheinlichkeiten zur Sicherheit begrenzen
max_hazard_rate = 1.0
max_reentry_quote = 0.9

# WICHTIG:
# True, wenn bestand_df["Berichtsjahr"] bereits vorher so angepasst wurde,
# dass z.B. aus 01.01.2026 das Jahr 2025 wurde.
bestand_berichtsjahr_ist_bereits_angepasst = True

# Benötigte Input-DataFrames:
# bestand_df          = Bestand
# nzl_df              = Neuzulassungen
# hazard_pooled_df    = gepoolte Ausscheidewahrscheinlichkeiten
# reentry_pooled_df   = gepoolte Reentry-Quoten
# fleet_total_df      = Ziel-Gesamtbestand je Jahr


# ------------------------------------------------------------
# Hilfsfunktionen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr", bestand_logik=False):
    """
    Wandelt Berichtsjahr robust in ein Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Wenn bestand_logik=True:
    Ein Bestand mit Datum 01.01.2026 wird als Bestand des Jahres 2025 interpretiert.
    Deshalb wird dann minus 1 gerechnet.

    Wenn bestand_df["Berichtsjahr"] bereits vorher angepasst wurde,
    muss bestand_logik=False verwendet werden.
    """
    s = df[spalte]

    jahr_num = pd.to_numeric(s, errors="coerce")
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    jahr = jahr_num.fillna(jahr_date)

    if bestand_logik:
        jahr = jahr - 1

    return jahr.astype("Int64")


def berechne_altersklasse(df, alter_spalte="Alter", breite=5):
    """
    Erstellt Altersklasse_Start, Altersklasse_Ende und Altersklasse.
    """
    df["Altersklasse_Start"] = (df[alter_spalte] // breite) * breite
    df["Altersklasse_Ende"] = df["Altersklasse_Start"] + breite - 1

    df["Altersklasse"] = (
        df["Altersklasse_Start"].astype(int).astype(str)
        + "-"
        + df["Altersklasse_Ende"].astype(int).astype(str)
    )

    return df


# ------------------------------------------------------------
# 1. Bestand 2025 bereinigen und aggregieren
# ------------------------------------------------------------

bestand_work = bestand_df.copy()

bestand_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung",
    "Anzahl"
]

bestand_work = bestand_work[bestand_relevante_spalten].copy()

# KORRIGIERTE VERSION:
# Wenn Berichtsjahr im Bestand bereits angepasst ist, wird NICHT erneut -1 gerechnet.
bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
    bestand_work,
    "Berichtsjahr",
    bestand_logik=not bestand_berichtsjahr_ist_bereits_angepasst
)

bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
    bestand_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")

bestand_work["Anzahl"] = pd.to_numeric(
    bestand_work["Anzahl"],
    errors="coerce"
)

bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

bestand_2025 = bestand_work[
    bestand_work["Berichtsjahr"] == basisjahr
].copy()

print("Gefundene Zeilen für Bestand 2025:", len(bestand_2025))
print("Summe Bestand 2025:", bestand_2025["Anzahl"].sum())

if bestand_2025.empty:
    raise ValueError(
        f"Für das Basisjahr {basisjahr} wurde kein Bestand gefunden. "
        "Prüfe, ob bestand_df['Berichtsjahr'] bereits angepasst wurde "
        "und ob die Einstellung bestand_berichtsjahr_ist_bereits_angepasst korrekt ist."
    )

bestand_2025 = (
    bestand_2025
    .groupby(
        ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
        as_index=False
    )["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand_2025"})
)

# Alter im Basisjahr berechnen
bestand_2025["Alter"] = (
    basisjahr - bestand_2025["Jahr der Erstzulassung"]
)

bestand_2025 = bestand_2025[
    bestand_2025["Alter"] >= 0
].copy()

bestand_2025 = berechne_altersklasse(
    bestand_2025,
    alter_spalte="Alter",
    breite=altersklassen_breite
)


# ------------------------------------------------------------
# 2. Ausscheidewahrscheinlichkeiten vorbereiten
# ------------------------------------------------------------

hazard_rates = hazard_pooled_df.copy()

hazard_rates["pooled_hazard_rate"] = pd.to_numeric(
    hazard_rates["pooled_hazard_rate"],
    errors="coerce"
).fillna(0)

hazard_rates["pooled_hazard_rate"] = (
    hazard_rates["pooled_hazard_rate"]
    .clip(lower=0, upper=max_hazard_rate)
)

hazard_rates = hazard_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_hazard_rate"
    ]
].copy()


# ------------------------------------------------------------
# 3. Reentry-Quoten vorbereiten
# ------------------------------------------------------------

reentry_rates = reentry_pooled_df.copy()

reentry_rates["pooled_reentry_quote_exits"] = pd.to_numeric(
    reentry_rates["pooled_reentry_quote_exits"],
    errors="coerce"
).fillna(0)

reentry_rates["pooled_reentry_quote_exits"] = (
    reentry_rates["pooled_reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_rates = reentry_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_reentry_quote_exits"
    ]
].copy()


# ------------------------------------------------------------
# 4. Bestand 2025 mit Hazard-Rate und Reentry-Quote verbinden
# ------------------------------------------------------------

projection_base = bestand_2025.merge(
    hazard_rates,
    on=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende"
    ],
    how="left"
)

projection_base = projection_base.merge(
    reentry_rates,
    on=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende"
    ],
    how="left"
)

# Fehlende Wahrscheinlichkeiten als 0 behandeln
projection_base["pooled_hazard_rate"] = (
    projection_base["pooled_hazard_rate"]
    .fillna(0)
    .clip(lower=0, upper=max_hazard_rate)
)

projection_base["pooled_reentry_quote_exits"] = (
    projection_base["pooled_reentry_quote_exits"]
    .fillna(0)
    .clip(lower=0, upper=max_reentry_quote)
)


# ------------------------------------------------------------
# 5. Exits 2026 und Reentries 2026 berechnen
# ------------------------------------------------------------

projection_base["exits_2026"] = (
    projection_base["bestand_2025"]
    * projection_base["pooled_hazard_rate"]
)

projection_base["reentries_2026"] = (
    projection_base["exits_2026"]
    * projection_base["pooled_reentry_quote_exits"]
)

projection_base["bestand_nach_exits_reentries_2026"] = (
    projection_base["bestand_2025"]
    - projection_base["exits_2026"]
    + projection_base["reentries_2026"]
)

projection_base["bestand_nach_exits_reentries_2026"] = (
    projection_base["bestand_nach_exits_reentries_2026"]
    .clip(lower=0)
)


# ------------------------------------------------------------
# 6. Ziel-Gesamtbestand 2026 aus fleet_total_df holen
# ------------------------------------------------------------

fleet_total_work = fleet_total_df.copy()

fleet_total_work["Jahr"] = pd.to_numeric(
    fleet_total_work["Jahr"],
    errors="coerce"
).astype("Int64")

fleet_total_work["gesamtbestand_target"] = pd.to_numeric(
    fleet_total_work["gesamtbestand_target"],
    errors="coerce"
)

zielbestand_2026_series = fleet_total_work.loc[
    fleet_total_work["Jahr"] == prognosejahr,
    "gesamtbestand_target"
]

if zielbestand_2026_series.empty:
    raise ValueError(
        f"Für das Prognosejahr {prognosejahr} wurde kein Zielbestand "
        "in fleet_total_df gefunden."
    )

zielbestand_2026 = zielbestand_2026_series.iloc[0]

bestand_nach_exits_reentries = (
    projection_base["bestand_nach_exits_reentries_2026"].sum()
)

fehlende_fahrzeuge = max(
    zielbestand_2026 - bestand_nach_exits_reentries,
    0
)


# ------------------------------------------------------------
# 7. Verteilung der neuen Fahrzeuge nach NZL-Verteilung 2025
# ------------------------------------------------------------

nzl_work = nzl_df.copy()

nzl_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Anzahl"
]

nzl_work = nzl_work[nzl_relevante_spalten].copy()

nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(
    nzl_work,
    "Berichtsjahr",
    bestand_logik=False
)

nzl_work["Anzahl"] = pd.to_numeric(
    nzl_work["Anzahl"],
    errors="coerce"
)

nzl_work = nzl_work.dropna(subset=nzl_relevante_spalten)

nzl_2025 = nzl_work[
    nzl_work["Berichtsjahr"] == basisjahr
].copy()

nzl_verteilung = (
    nzl_2025
    .groupby(["Segment", "Antriebsart"], as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "nzl_2025"})
)

summe_nzl_2025 = nzl_verteilung["nzl_2025"].sum()

if summe_nzl_2025 > 0:
    nzl_verteilung["anteil_nzl_2025"] = (
        nzl_verteilung["nzl_2025"] / summe_nzl_2025
    )
else:
    raise ValueError(
        f"Für das Jahr {basisjahr} wurden keine Neuzulassungen gefunden. "
        "Die Verteilung neuer Fahrzeuge kann daher nicht berechnet werden."
    )

nzl_verteilung["neuzulassungen_2026"] = (
    fehlende_fahrzeuge * nzl_verteilung["anteil_nzl_2025"]
)

# Neue Fahrzeuge des Jahres 2026 werden als Kohorte Erstzulassungsjahr 2026 angelegt
new_vehicles_2026 = nzl_verteilung[
    ["Segment", "Antriebsart", "neuzulassungen_2026"]
].copy()

new_vehicles_2026["Jahr der Erstzulassung"] = prognosejahr

new_vehicles_2026 = new_vehicles_2026.rename(
    columns={"neuzulassungen_2026": "bestand_2026"}
)


# ------------------------------------------------------------
# 8. Fortgeschriebenen Bestand 2026 aus Altbestand + neuen Fahrzeugen erstellen
# ------------------------------------------------------------

bestand_alt_2026 = projection_base[
    [
        "Segment",
        "Antriebsart",
        "Jahr der Erstzulassung",
        "bestand_nach_exits_reentries_2026"
    ]
].copy()

bestand_alt_2026 = bestand_alt_2026.rename(
    columns={"bestand_nach_exits_reentries_2026": "bestand_2026"}
)

bestand_2026_df = pd.concat(
    [
        bestand_alt_2026,
        new_vehicles_2026
    ],
    ignore_index=True
)

bestand_2026_df = (
    bestand_2026_df
    .groupby(
        ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
        as_index=False
    )["bestand_2026"]
    .sum()
)

bestand_2026_df["Berichtsjahr"] = prognosejahr

bestand_2026_df = bestand_2026_df[
    [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Jahr der Erstzulassung",
        "bestand_2026"
    ]
].copy()


# ------------------------------------------------------------
# 9. Kontrolle ausgeben
# ------------------------------------------------------------

finaler_bestand_2026 = bestand_2026_df["bestand_2026"].sum()

print(f"Zielbestand {prognosejahr}: {zielbestand_2026:,.0f}")
print(
    f"Bestand nach Abgängen und Rückkehrern {prognosejahr}: "
    f"{bestand_nach_exits_reentries:,.0f}"
)
print(f"Fehlende Fahrzeuge / Neuzulassungen {prognosejahr}: {fehlende_fahrzeuge:,.0f}")
print(f"Finaler Bestand {prognosejahr}: {finaler_bestand_2026:,.0f}")

print("\nSumme erwartete Exits 2026:")
print(f"{projection_base['exits_2026'].sum():,.0f}")

print("\nSumme erwartete Reentries 2026:")
print(f"{projection_base['reentries_2026'].sum():,.0f}")

print("\nFortgeschriebener Bestand 2026:")
display(bestand_2026_df.head())

print("\nKontrolltabelle je Kohorte:")
display(projection_base.head())
```

# Fortschreibung bis 2050

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

basisjahr = 2025
zieljahr = 2050

altersklassen_breite = 5

# Wahrscheinlichkeiten sinnvoll begrenzen
max_hazard_rate = 1.0
max_reentry_quote = 0.9

# WICHTIG:
# True, wenn bestand_df["Berichtsjahr"] bereits vorher so angepasst wurde,
# dass z.B. aus 01.01.2026 das Jahr 2025 wurde.
bestand_berichtsjahr_ist_bereits_angepasst = True

# Benötigte Input-DataFrames:
# bestand_df          = Bestand
# nzl_df              = Neuzulassungen
# hazard_pooled_df    = gepoolte Ausscheidewahrscheinlichkeiten
# reentry_pooled_df   = gepoolte Reentry-Quoten
# fleet_total_df      = Ziel-Gesamtbestand je Jahr


# ------------------------------------------------------------
# Hilfsfunktionen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr", bestand_logik=False):
    """
    Wandelt Berichtsjahr robust in ein Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Wenn bestand_logik=True:
    Ein Bestand mit Datum 01.01.2026 wird als Bestand des Jahres 2025 interpretiert.
    Deshalb wird dann minus 1 gerechnet.
    """
    s = df[spalte]

    jahr_num = pd.to_numeric(s, errors="coerce")
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    jahr = jahr_num.fillna(jahr_date)

    if bestand_logik:
        jahr = jahr - 1

    return jahr.astype("Int64")


def berechne_altersklasse(df, aktuelles_jahr, breite=5):
    """
    Berechnet Alter und 5-Jahres-Altersklasse für den aktuellen Bestand.
    """
    df["Alter"] = aktuelles_jahr - df["Jahr der Erstzulassung"]

    df = df[df["Alter"] >= 0].copy()

    df["Altersklasse_Start"] = (df["Alter"] // breite) * breite
    df["Altersklasse_Ende"] = df["Altersklasse_Start"] + breite - 1

    df["Altersklasse"] = (
        df["Altersklasse_Start"].astype(int).astype(str)
        + "-"
        + df["Altersklasse_Ende"].astype(int).astype(str)
    )

    return df


def safe_sum(series):
    """
    Summiert robust und behandelt fehlende Werte als 0.
    """
    return pd.to_numeric(series, errors="coerce").fillna(0).sum()


# ------------------------------------------------------------
# 1. Ausscheidewahrscheinlichkeiten vorbereiten
# ------------------------------------------------------------

hazard_rates = hazard_pooled_df.copy()

hazard_rates["pooled_hazard_rate"] = pd.to_numeric(
    hazard_rates["pooled_hazard_rate"],
    errors="coerce"
).fillna(0)

hazard_rates["pooled_hazard_rate"] = (
    hazard_rates["pooled_hazard_rate"]
    .clip(lower=0, upper=max_hazard_rate)
)

hazard_rates = hazard_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_hazard_rate"
    ]
].copy()

hazard_rates["Altersklasse_Start"] = pd.to_numeric(
    hazard_rates["Altersklasse_Start"],
    errors="coerce"
).astype("Int64")

hazard_rates["Altersklasse_Ende"] = pd.to_numeric(
    hazard_rates["Altersklasse_Ende"],
    errors="coerce"
).astype("Int64")

hazard_rates = hazard_rates.dropna(
    subset=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_hazard_rate"
    ]
)


# ------------------------------------------------------------
# 2. Rückkehrquoten vorbereiten
# ------------------------------------------------------------

reentry_rates = reentry_pooled_df.copy()

reentry_rates["pooled_reentry_quote_exits"] = pd.to_numeric(
    reentry_rates["pooled_reentry_quote_exits"],
    errors="coerce"
).fillna(0)

reentry_rates["pooled_reentry_quote_exits"] = (
    reentry_rates["pooled_reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_rates = reentry_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_reentry_quote_exits"
    ]
].copy()

reentry_rates["Altersklasse_Start"] = pd.to_numeric(
    reentry_rates["Altersklasse_Start"],
    errors="coerce"
).astype("Int64")

reentry_rates["Altersklasse_Ende"] = pd.to_numeric(
    reentry_rates["Altersklasse_Ende"],
    errors="coerce"
).astype("Int64")

reentry_rates = reentry_rates.dropna(
    subset=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_reentry_quote_exits"
    ]
)


# ------------------------------------------------------------
# 3. Feste Neuzulassungsverteilung aus 2025 berechnen
# ------------------------------------------------------------

nzl_work = nzl_df.copy()

nzl_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Anzahl"
]

nzl_work = nzl_work[nzl_relevante_spalten].copy()

nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(
    nzl_work,
    "Berichtsjahr",
    bestand_logik=False
)

nzl_work["Anzahl"] = pd.to_numeric(
    nzl_work["Anzahl"],
    errors="coerce"
)

nzl_work = nzl_work.dropna(subset=nzl_relevante_spalten)

nzl_2025 = nzl_work[
    nzl_work["Berichtsjahr"] == basisjahr
].copy()

nzl_verteilung = (
    nzl_2025
    .groupby(["Segment", "Antriebsart"], as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "nzl_2025"})
)

summe_nzl_2025 = safe_sum(nzl_verteilung["nzl_2025"])

if summe_nzl_2025 <= 0:
    raise ValueError(
        f"Für das Jahr {basisjahr} wurden keine Neuzulassungen gefunden. "
        "Die Verteilung neuer Fahrzeuge kann daher nicht berechnet werden."
    )

nzl_verteilung["anteil_nzl_2025"] = (
    nzl_verteilung["nzl_2025"] / summe_nzl_2025
)

nzl_verteilung = nzl_verteilung[
    ["Segment", "Antriebsart", "anteil_nzl_2025"]
].copy()


# ------------------------------------------------------------
# 4. Ziel-Gesamtbestände vorbereiten
# ------------------------------------------------------------

fleet_total_work = fleet_total_df.copy()

fleet_total_work["Jahr"] = pd.to_numeric(
    fleet_total_work["Jahr"],
    errors="coerce"
).astype("Int64")

fleet_total_work["gesamtbestand_target"] = pd.to_numeric(
    fleet_total_work["gesamtbestand_target"],
    errors="coerce"
)

fleet_total_work = fleet_total_work.dropna(
    subset=["Jahr", "gesamtbestand_target"]
)

target_dict = dict(
    zip(
        fleet_total_work["Jahr"].astype(int),
        fleet_total_work["gesamtbestand_target"]
    )
)

fehlende_zieljahre = [
    jahr for jahr in range(basisjahr, zieljahr + 1)
    if jahr not in target_dict
]

if fehlende_zieljahre:
    raise ValueError(
        "Für folgende Jahre fehlen Zielbestände in fleet_total_df: "
        + ", ".join(map(str, fehlende_zieljahre))
    )


# ------------------------------------------------------------
# 5. Startbestand 2025 vorbereiten
# ------------------------------------------------------------

bestand_work = bestand_df.copy()

bestand_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung",
    "Anzahl"
]

bestand_work = bestand_work[bestand_relevante_spalten].copy()

bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
    bestand_work,
    "Berichtsjahr",
    bestand_logik=not bestand_berichtsjahr_ist_bereits_angepasst
)

bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
    bestand_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")

bestand_work["Anzahl"] = pd.to_numeric(
    bestand_work["Anzahl"],
    errors="coerce"
)

bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

current_stock = bestand_work[
    bestand_work["Berichtsjahr"] == basisjahr
].copy()

if current_stock.empty:
    raise ValueError(
        f"Für das Basisjahr {basisjahr} wurde kein Bestand gefunden. "
        "Prüfe die Spalte Berichtsjahr und die Einstellung "
        "bestand_berichtsjahr_ist_bereits_angepasst."
    )

current_stock = (
    current_stock
    .groupby(
        ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
        as_index=False
    )["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand"})
)

current_stock["Jahr der Erstzulassung"] = current_stock[
    "Jahr der Erstzulassung"
].astype(int)

current_stock["bestand"] = pd.to_numeric(
    current_stock["bestand"],
    errors="coerce"
).fillna(0)

current_stock = current_stock[current_stock["bestand"] > 0].copy()


# ------------------------------------------------------------
# 6. Mehrjährige Fortschreibung 2025 bis 2050
# ------------------------------------------------------------

results_by_year = {}

overview_rows = []

# Optional: Startjahr separat speichern
results_by_year[basisjahr] = {
    "finaler_bestand": current_stock.assign(Berichtsjahr=basisjahr).copy()
}

overview_rows.append({
    "Jahr": basisjahr,
    "zielbestand": target_dict[basisjahr],
    "bestand_vor_neuen_fahrzeugen": safe_sum(current_stock["bestand"]),
    "fehlende_fahrzeuge": max(
        target_dict[basisjahr] - safe_sum(current_stock["bestand"]),
        0
    ),
    "finaler_bestand": safe_sum(current_stock["bestand"])
})


for jahr in range(basisjahr, zieljahr):

    folgejahr = jahr + 1

    # --------------------------------------------------------
    # 6.1 Alter und Altersklassen im aktuellen Jahr berechnen
    # --------------------------------------------------------

    projection_base = current_stock.copy()

    projection_base = berechne_altersklasse(
        projection_base,
        aktuelles_jahr=jahr,
        breite=altersklassen_breite
    )

    # --------------------------------------------------------
    # 6.2 Hazard-Raten und Reentry-Quoten ergänzen
    # --------------------------------------------------------

    projection_base = projection_base.merge(
        hazard_rates,
        on=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende"
        ],
        how="left"
    )

    projection_base = projection_base.merge(
        reentry_rates,
        on=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende"
        ],
        how="left"
    )

    projection_base["pooled_hazard_rate"] = (
        projection_base["pooled_hazard_rate"]
        .fillna(0)
        .clip(lower=0, upper=max_hazard_rate)
    )

    projection_base["pooled_reentry_quote_exits"] = (
        projection_base["pooled_reentry_quote_exits"]
        .fillna(0)
        .clip(lower=0, upper=max_reentry_quote)
    )

    # --------------------------------------------------------
    # 6.3 Exits und Reentries für das Folgejahr berechnen
    # --------------------------------------------------------

    projection_base[f"exits_{folgejahr}"] = (
        projection_base["bestand"]
        * projection_base["pooled_hazard_rate"]
    )

    projection_base[f"reentries_{folgejahr}"] = (
        projection_base[f"exits_{folgejahr}"]
        * projection_base["pooled_reentry_quote_exits"]
    )

    projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"] = (
        projection_base["bestand"]
        - projection_base[f"exits_{folgejahr}"]
        + projection_base[f"reentries_{folgejahr}"]
    )

    projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"] = (
        projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"]
        .clip(lower=0)
    )

    bestehender_bestand_folgejahr = safe_sum(
        projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"]
    )

    zielbestand_folgejahr = target_dict[folgejahr]

    fehlende_fahrzeuge = max(
        zielbestand_folgejahr - bestehender_bestand_folgejahr,
        0
    )

    # --------------------------------------------------------
    # 6.4 Neue Fahrzeuge auf Basis der NZL-Verteilung 2025 ergänzen
    # --------------------------------------------------------

    new_vehicles = nzl_verteilung.copy()

    new_vehicles["Jahr der Erstzulassung"] = folgejahr

    new_vehicles["neue_fahrzeuge"] = (
        fehlende_fahrzeuge * new_vehicles["anteil_nzl_2025"]
    )

    new_vehicles = new_vehicles[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            "neue_fahrzeuge"
        ]
    ].copy()

    # --------------------------------------------------------
    # 6.5 Finalen Bestand des Folgejahres erstellen
    # --------------------------------------------------------

    stock_existing_next = projection_base[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            f"bestand_vor_neuen_fahrzeugen_{folgejahr}"
        ]
    ].copy()

    stock_existing_next = stock_existing_next.rename(
        columns={
            f"bestand_vor_neuen_fahrzeugen_{folgejahr}": "bestand"
        }
    )

    stock_new_next = new_vehicles.rename(
        columns={"neue_fahrzeuge": "bestand"}
    )

    current_stock_next = pd.concat(
        [
            stock_existing_next,
            stock_new_next
        ],
        ignore_index=True
    )

    current_stock_next = (
        current_stock_next
        .groupby(
            ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
            as_index=False
        )["bestand"]
        .sum()
    )

    current_stock_next["bestand"] = pd.to_numeric(
        current_stock_next["bestand"],
        errors="coerce"
    ).fillna(0)

    current_stock_next = current_stock_next[
        current_stock_next["bestand"] > 0
    ].copy()

    finaler_bestand_folgejahr = safe_sum(current_stock_next["bestand"])

    # --------------------------------------------------------
    # 6.6 Ergebnisse für dieses Jahr speichern
    # --------------------------------------------------------

    exits_df = projection_base[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            f"exits_{folgejahr}"
        ]
    ].copy().rename(
        columns={f"exits_{folgejahr}": "exits"}
    )

    exits_df["Berichtsjahr"] = folgejahr

    reentries_df = projection_base[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            f"reentries_{folgejahr}"
        ]
    ].copy().rename(
        columns={f"reentries_{folgejahr}": "reentries"}
    )

    reentries_df["Berichtsjahr"] = folgejahr

    bestand_vor_neu_df = stock_existing_next.copy()
    bestand_vor_neu_df["Berichtsjahr"] = folgejahr
    bestand_vor_neu_df = bestand_vor_neu_df.rename(
        columns={"bestand": "bestand_vor_neuen_fahrzeugen"}
    )

    neue_fahrzeuge_df = new_vehicles.copy()
    neue_fahrzeuge_df["Berichtsjahr"] = folgejahr

    finaler_bestand_df = current_stock_next.copy()
    finaler_bestand_df["Berichtsjahr"] = folgejahr
    finaler_bestand_df = finaler_bestand_df.rename(
        columns={"bestand": "finaler_bestand"}
    )

    results_by_year[folgejahr] = {
        "abgaenge": exits_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "exits"
            ]
        ].copy(),

        "rueckkehrer": reentries_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "reentries"
            ]
        ].copy(),

        "bestand_vor_neuen_fahrzeugen": bestand_vor_neu_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "bestand_vor_neuen_fahrzeugen"
            ]
        ].copy(),

        "fehlende_fahrzeuge": pd.DataFrame({
            "Berichtsjahr": [folgejahr],
            "fehlende_fahrzeuge": [fehlende_fahrzeuge]
        }),

        "neue_fahrzeuge": neue_fahrzeuge_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "neue_fahrzeuge"
            ]
        ].copy(),

        "finaler_bestand": finaler_bestand_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "finaler_bestand"
            ]
        ].copy()
    }

    overview_rows.append({
        "Jahr": folgejahr,
        "zielbestand": zielbestand_folgejahr,
        "bestand_vor_neuen_fahrzeugen": bestehender_bestand_folgejahr,
        "fehlende_fahrzeuge": fehlende_fahrzeuge,
        "finaler_bestand": finaler_bestand_folgejahr
    })

    # Finaler Bestand des Folgejahres wird Startbestand für das nächste Jahr
    current_stock = current_stock_next.copy()


# ------------------------------------------------------------
# 7. Kompakte Übersichtstabelle erstellen
# ------------------------------------------------------------

fleet_projection_overview_df = pd.DataFrame(overview_rows)

fleet_projection_overview_df["abweichung_final_vs_ziel"] = (
    fleet_projection_overview_df["finaler_bestand"]
    - fleet_projection_overview_df["zielbestand"]
)

fleet_projection_overview_df = fleet_projection_overview_df[
    [
        "Jahr",
        "zielbestand",
        "bestand_vor_neuen_fahrzeugen",
        "fehlende_fahrzeuge",
        "finaler_bestand",
        "abweichung_final_vs_ziel"
    ]
].copy()


# ------------------------------------------------------------
# 8. Kontrolle anzeigen
# ------------------------------------------------------------

print(f"Fortschreibung von {basisjahr} bis {zieljahr}")
print(f"Startbestand {basisjahr}: {safe_sum(results_by_year[basisjahr]['finaler_bestand']['bestand']):,.0f}")
print(f"Zielbestand {zieljahr}: {target_dict[zieljahr]:,.0f}")
print(f"Finaler Bestand {zieljahr}: {fleet_projection_overview_df.loc[fleet_projection_overview_df['Jahr'] == zieljahr, 'finaler_bestand'].iloc[0]:,.0f}")

print("\nÜbersicht:")
display(fleet_projection_overview_df.head())
display(fleet_projection_overview_df.tail())

print("\nBeispiel: Ergebnisse für 2026")
display(results_by_year[2026]["finaler_bestand"].head())
display(results_by_year[2026]["abgaenge"].head())
display(results_by_year[2026]["rueckkehrer"].head())
display(results_by_year[2026]["neue_fahrzeuge"].head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Bestand pro Jahr und Antriebsart aus results_by_year sammeln
# ------------------------------------------------------------

stock_by_drive_rows = []

for jahr, year_results in results_by_year.items():

    # Für das Basisjahr kann die Bestandsspalte ggf. "bestand" heißen,
    # für Folgejahre aus dem Fortschreibungscode "finaler_bestand".
    stock_df = year_results["finaler_bestand"].copy()

    if "finaler_bestand" in stock_df.columns:
        bestand_spalte = "finaler_bestand"
    elif "bestand" in stock_df.columns:
        bestand_spalte = "bestand"
    else:
        raise ValueError(
            f"Für das Jahr {jahr} wurde keine Bestandsspalte gefunden. "
            "Erwartet wurde 'finaler_bestand' oder 'bestand'."
        )

    stock_df[bestand_spalte] = pd.to_numeric(
        stock_df[bestand_spalte],
        errors="coerce"
    ).fillna(0)

    stock_by_drive = (
        stock_df
        .groupby("Antriebsart", as_index=False)[bestand_spalte]
        .sum()
        .rename(columns={bestand_spalte: "fahrzeugbestand"})
    )

    stock_by_drive["Jahr"] = jahr

    stock_by_drive_rows.append(stock_by_drive)


# Gemeinsames DataFrame erstellen
stock_by_drive_year_df = pd.concat(
    stock_by_drive_rows,
    ignore_index=True
)


# ------------------------------------------------------------
# 2. Pivot-Tabelle erstellen
# ------------------------------------------------------------

stock_by_drive_pivot = (
    stock_by_drive_year_df
    .pivot_table(
        index="Jahr",
        columns="Antriebsart",
        values="fahrzeugbestand",
        aggfunc="sum",
        fill_value=0
    )
    .sort_index()
)


# ------------------------------------------------------------
# 3. Gestapeltes Balkendiagramm erstellen
# ------------------------------------------------------------

ax = stock_by_drive_pivot.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 7)
)

ax.set_xlabel("Year")
ax.set_ylabel("Vehicle stock")
ax.set_title("Vehicle stock by drive type")

# Legende rechts außerhalb der Grafik platzieren
ax.legend(
    title="Drive type",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 4. Kontrolle: erste Zeilen der Pivot-Tabelle anzeigen
# ------------------------------------------------------------

display(stock_by_drive_pivot.head())
```
