# Fleet Model

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

basisjahr = 2025
zieljahr = 2050

# Prozentuale Veränderung gegenüber dem Basisbestand
# Beispiele:
# +10 %  ->  10
# -20 %  -> -20
veraenderung_prozent = 100

# Art der Fortschreibung:
# "linear"  = jedes Jahr gleicher absoluter Zuwachs/Rückgang
# "growth"  = jedes Jahr gleiche prozentuale Wachstumsrate
fortschreibung = "growth"


# ------------------------------------------------------------
# 2. Gesamtbestand je Jahr berechnen
# ------------------------------------------------------------

bestand_total_by_year = (
    bestand_df
    .groupby("Berichtsjahr", as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Berichtsjahr": "Jahr", "Anzahl": "gesamtbestand"})
)


# ------------------------------------------------------------
# 3. Basisbestand bestimmen
# ------------------------------------------------------------

basisbestand = bestand_total_by_year.loc[
    bestand_total_by_year["Jahr"] == basisjahr,
    "gesamtbestand"
].iloc[0]


# ------------------------------------------------------------
# 4. Zielbestand berechnen
# ------------------------------------------------------------

zielbestand = basisbestand * (1 + veraenderung_prozent / 100)


# ------------------------------------------------------------
# 5. Zeitreihe vom Basisjahr bis Zieljahr erzeugen
# ------------------------------------------------------------

jahre = np.arange(basisjahr, zieljahr + 1)
anzahl_jahre = zieljahr - basisjahr

if fortschreibung == "linear":
    # Jedes Jahr gleicher absoluter Zuwachs/Rückgang
    gesamtbestand_target = np.linspace(
        basisbestand,
        zielbestand,
        len(jahre)
    )

elif fortschreibung == "growth":
    # Jedes Jahr gleiche prozentuale Wachstumsrate
    wachstumsrate = (zielbestand / basisbestand) ** (1 / anzahl_jahre) - 1

    gesamtbestand_target = basisbestand * (
        (1 + wachstumsrate) ** np.arange(len(jahre))
    )

else:
    raise ValueError("fortschreibung muss entweder 'linear' oder 'growth' sein.")


# ------------------------------------------------------------
# 6. Output-DataFrame erstellen
# ------------------------------------------------------------

fleet_total_df = pd.DataFrame({
    "Jahr": jahre,
    "gesamtbestand_target": gesamtbestand_target
})


# ------------------------------------------------------------
# 7. Kontrolle ausgeben
# ------------------------------------------------------------

print("Basisjahr:", basisjahr)
print("Basisbestand:", basisbestand)

print("Zieljahr:", zieljahr)
print("Zielbestand:", zielbestand)

print("Fortschreibung:", fortschreibung)

print("\nErste Zeilen:")
display(fleet_total_df.head())

print("\nLetzte Zeilen:")
display(fleet_total_df.tail())
```

```python
import matplotlib.pyplot as plt

# Liniendiagramm für die Entwicklung des Gesamtbestands
ax = fleet_total_df.plot(
    x="Jahr",
    y="gesamtbestand_target",
    kind="line",
    marker="o",
    figsize=(12, 6),
    legend=False
)

ax.set_xlabel("Year")
ax.set_ylabel("Total stock")
ax.set_title("Development of total vehicle stock")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

anzahl_betrachtete_jahre = 5
altersklassen_breite = 5

gruppen_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung"
]


# ------------------------------------------------------------
# 1. Arbeitskopien erstellen
# ------------------------------------------------------------

abs_work = abs_df.copy()
bestand_work = bestand_df.copy()


# ------------------------------------------------------------
# 2. Relevante Spalten prüfen und bereinigen
# ------------------------------------------------------------

relevante_spalten = gruppen_spalten + ["Anzahl"]

abs_work = abs_work[relevante_spalten].copy()
bestand_work = bestand_work[relevante_spalten].copy()

# Zahlen-/Jahresspalten sicher numerisch machen
for df in [abs_work, bestand_work]:
    df["Berichtsjahr"] = pd.to_numeric(df["Berichtsjahr"], errors="coerce")
    df["Jahr der Erstzulassung"] = pd.to_numeric(
        df["Jahr der Erstzulassung"],
        errors="coerce"
    )
    df["Anzahl"] = pd.to_numeric(df["Anzahl"], errors="coerce")

# Fehlende oder ungültige Werte ausschließen
abs_work = abs_work.dropna(subset=relevante_spalten)
bestand_work = bestand_work.dropna(subset=relevante_spalten)

# Jahre als Integer speichern
for df in [abs_work, bestand_work]:
    df["Berichtsjahr"] = df["Berichtsjahr"].astype(int)
    df["Jahr der Erstzulassung"] = df["Jahr der Erstzulassung"].astype(int)


# ------------------------------------------------------------
# 3. Letzte verfügbare Jahre der Abgangsdaten bestimmen
# ------------------------------------------------------------

aktuelles_jahr = abs_work["Berichtsjahr"].max()
startjahr = aktuelles_jahr - anzahl_betrachtete_jahre + 1

basisjahre = list(range(startjahr, aktuelles_jahr + 1))

print(f"Basisjahre: {startjahr}-{aktuelles_jahr}")


# ------------------------------------------------------------
# 4. Nur betrachtete Jahre verwenden
# ------------------------------------------------------------

abs_work = abs_work[abs_work["Berichtsjahr"].isin(basisjahre)].copy()
bestand_work = bestand_work[bestand_work["Berichtsjahr"].isin(basisjahre)].copy()


# ------------------------------------------------------------
# 5. Abgänge und Bestände auf gemeinsame Gruppierung aggregieren
# ------------------------------------------------------------

abs_grouped = (
    abs_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "abgaenge"})
)

bestand_grouped = (
    bestand_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand"})
)


# ------------------------------------------------------------
# 6. Abgänge mit passendem Bestand verbinden
# ------------------------------------------------------------

hazard_base = abs_grouped.merge(
    bestand_grouped,
    on=gruppen_spalten,
    how="inner"
)


# ------------------------------------------------------------
# 7. Fahrzeugalter berechnen und ungültige Fälle ausschließen
# ------------------------------------------------------------

hazard_base["Alter"] = (
    hazard_base["Berichtsjahr"] - hazard_base["Jahr der Erstzulassung"]
)

hazard_base = hazard_base[
    (hazard_base["bestand"].notna()) &
    (hazard_base["bestand"] > 0) &
    (hazard_base["Alter"] >= 0)
].copy()


# ------------------------------------------------------------
# 8. Fahrzeugalter in 5-Jahres-Altersklassen einteilen
# ------------------------------------------------------------

hazard_base["Altersklasse_Start"] = (
    hazard_base["Alter"] // altersklassen_breite
) * altersklassen_breite

hazard_base["Altersklasse_Ende"] = (
    hazard_base["Altersklasse_Start"] + altersklassen_breite - 1
)

hazard_base["Altersklasse"] = (
    hazard_base["Altersklasse_Start"].astype(str)
    + "-"
    + hazard_base["Altersklasse_Ende"].astype(str)
)


# ------------------------------------------------------------
# 9. Gepoolte Hazard-Rate berechnen
# ------------------------------------------------------------

hazard_pooled_df = (
    hazard_base
    .groupby(
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "Altersklasse"
        ],
        as_index=False
    )
    .agg(
        summe_abgaenge=("abgaenge", "sum"),
        summe_bestand=("bestand", "sum"),
        anzahl_betrachtete_jahre=("Berichtsjahr", "nunique")
    )
)

hazard_pooled_df["pooled_hazard_rate"] = (
    hazard_pooled_df["summe_abgaenge"] / hazard_pooled_df["summe_bestand"]
)


# ------------------------------------------------------------
# 10. Optional: Sortierung
# ------------------------------------------------------------

hazard_pooled_df = hazard_pooled_df.sort_values(
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start"
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 11. Kontrolle anzeigen
# ------------------------------------------------------------

print(f"Basisjahre: {startjahr}-{aktuelles_jahr}")
print(f"Anzahl betrachtete Jahre: {anzahl_betrachtete_jahre}")
print(f"Breite der Altersklassen: {altersklassen_breite} Jahre")

display(hazard_pooled_df.head())
display(hazard_pooled_df.tail())
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

anzahl_betrachtete_jahre = 5
altersklassen_breite = 5
max_reentry_quote = 0.9

# Nur Altersklassen bis 20-24 berücksichtigen
max_alter = 24

gruppen_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung"
]


# ------------------------------------------------------------
# Hilfsfunktion: Berichtsjahr bereinigen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr"):
    """
    Wandelt Berichtsjahr robust in ein numerisches Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Falls ein vollständiges Datum vorliegt, wird das Kalenderjahr extrahiert.
    Hinweis:
    Wenn dein Bestand mit 01.01.2026 das Basisjahr 2025 beschreibt,
    wird für bestand_df unten explizit minus 1 gerechnet.
    """
    s = df[spalte]

    # Erst als Zahl versuchen
    jahr_num = pd.to_numeric(s, errors="coerce")

    # Für nicht-numerische Werte als Datum versuchen
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    return jahr_num.fillna(jahr_date).astype("Int64")


# ------------------------------------------------------------
# 1. Arbeitskopien erstellen
# ------------------------------------------------------------

abs_work = abs_df.copy()
nzl_work = nzl_df.copy()
bestand_work = bestand_df.copy()


# ------------------------------------------------------------
# 2. Exits aus abs_df vorbereiten und aggregieren
# ------------------------------------------------------------

abs_relevante_spalten = gruppen_spalten + ["Anzahl"]

abs_work = abs_work[abs_relevante_spalten].copy()

abs_work["Berichtsjahr"] = bereinige_berichtsjahr(abs_work, "Berichtsjahr")
abs_work["Jahr der Erstzulassung"] = pd.to_numeric(
    abs_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")
abs_work["Anzahl"] = pd.to_numeric(abs_work["Anzahl"], errors="coerce")

abs_work = abs_work.dropna(subset=abs_relevante_spalten)

exits_grouped = (
    abs_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "exits"})
)


# ------------------------------------------------------------
# 3. Neuzulassungen aus nzl_df vorbereiten und aggregieren
# ------------------------------------------------------------

nzl_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Anzahl"
]

nzl_work = nzl_work[nzl_relevante_spalten].copy()

nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(nzl_work, "Berichtsjahr")
nzl_work["Anzahl"] = pd.to_numeric(nzl_work["Anzahl"], errors="coerce")

# Neuzulassungen im Jahr t werden als Kohorte Erstzulassungsjahr t interpretiert
nzl_work["Jahr der Erstzulassung"] = nzl_work["Berichtsjahr"]

nzl_work = nzl_work.dropna(subset=gruppen_spalten + ["Anzahl"])

entries_grouped = (
    nzl_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "entries"})
)


# ------------------------------------------------------------
# 4. Bestände aus bestand_df vorbereiten und aggregieren
# ------------------------------------------------------------

bestand_relevante_spalten = gruppen_spalten + ["Anzahl"]

bestand_work = bestand_work[bestand_relevante_spalten].copy()

bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
    bestand_work,
    "Berichtsjahr"
)

# WICHTIG:
# Falls bestand_df in "Berichtsjahr" Datumswerte wie 01.01.2026 enthält,
# beschreibt dieser Bestand das Basisjahr 2025.
# Deshalb wird beim Bestand minus 1 gerechnet.
bestand_work["Berichtsjahr"] = bestand_work["Berichtsjahr"] - 1

bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
    bestand_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")
bestand_work["Anzahl"] = pd.to_numeric(bestand_work["Anzahl"], errors="coerce")

bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

stock_grouped = (
    bestand_work
    .groupby(gruppen_spalten, as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand"})
)


# ------------------------------------------------------------
# 5. Bestand t und Bestand t+1 derselben Kohorte zusammenführen
# ------------------------------------------------------------

stock_t = stock_grouped.rename(columns={"bestand": "bestand_t"}).copy()

stock_t_plus_1 = stock_grouped.copy()
stock_t_plus_1["Berichtsjahr"] = stock_t_plus_1["Berichtsjahr"] - 1
stock_t_plus_1 = stock_t_plus_1.rename(columns={"bestand": "bestand_t_plus_1"})

balance_df = stock_t.merge(
    stock_t_plus_1,
    on=gruppen_spalten,
    how="inner"
)


# ------------------------------------------------------------
# 6. Exits und Entries im Jahr t ergänzen
# ------------------------------------------------------------

balance_df = balance_df.merge(
    exits_grouped,
    on=gruppen_spalten,
    how="left"
)

balance_df = balance_df.merge(
    entries_grouped,
    on=gruppen_spalten,
    how="left"
)

balance_df["exits"] = balance_df["exits"].fillna(0)
balance_df["entries"] = balance_df["entries"].fillna(0)


# ------------------------------------------------------------
# 7. Reentries aus Bestandsbilanz berechnen
# ------------------------------------------------------------

balance_df["reentries_raw"] = (
    balance_df["bestand_t_plus_1"]
    - balance_df["bestand_t"]
    + balance_df["exits"]
    - balance_df["entries"]
)

balance_df["reentry_raw_negativ"] = balance_df["reentries_raw"] < 0

# Für Modellierung negative Reentries auf 0 setzen
balance_df["reentries"] = balance_df["reentries_raw"].clip(lower=0)


# ------------------------------------------------------------
# 8. Fahrzeugalter berechnen und Altersklassen bilden
# ------------------------------------------------------------

balance_df["Alter"] = (
    balance_df["Berichtsjahr"] - balance_df["Jahr der Erstzulassung"]
)

balance_df = balance_df[
    (balance_df["Alter"] >= 0) &
    (balance_df["Alter"] <= max_alter)
].copy()

balance_df["Altersklasse_Start"] = (
    balance_df["Alter"] // altersklassen_breite
) * altersklassen_breite

balance_df["Altersklasse_Ende"] = (
    balance_df["Altersklasse_Start"] + altersklassen_breite - 1
)

balance_df["Altersklasse"] = (
    balance_df["Altersklasse_Start"].astype(str)
    + "-"
    + balance_df["Altersklasse_Ende"].astype(str)
)


# ------------------------------------------------------------
# 9. Erster Output:
#    Reentries nach Jahr, Segment, Antriebsart und Altersklasse
# ------------------------------------------------------------

reentry_yearly_df = (
    balance_df
    .groupby(
        [
            "Berichtsjahr",
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "Altersklasse"
        ],
        as_index=False
    )
    .agg(
        summe_exits=("exits", "sum"),
        summe_reentries=("reentries", "sum"),
        summe_reentries_raw=("reentries_raw", "sum"),
        anzahl_negative_reentry_raw=("reentry_raw_negativ", "sum")
    )
)

reentry_yearly_df["reentry_quote_exits"] = np.where(
    reentry_yearly_df["summe_exits"] > 0,
    reentry_yearly_df["summe_reentries"] / reentry_yearly_df["summe_exits"],
    0
)

reentry_yearly_df["reentry_quote_exits"] = (
    reentry_yearly_df["reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_yearly_df = reentry_yearly_df.sort_values(
    [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Altersklasse_Start"
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 10. Kalibrierungsfenster bestimmen
# ------------------------------------------------------------

aktuelles_jahr = reentry_yearly_df["Berichtsjahr"].max()
startjahr = aktuelles_jahr - anzahl_betrachtete_jahre + 1

kalibrierungsjahre = list(range(startjahr, aktuelles_jahr + 1))

print(f"Kalibrierungsfenster: {startjahr}-{aktuelles_jahr}")


# ------------------------------------------------------------
# 11. Zweiter Output:
#     gepoolte Reentry-Quote über die letzten Jahre
# ------------------------------------------------------------

reentry_calibration_base = reentry_yearly_df[
    reentry_yearly_df["Berichtsjahr"].isin(kalibrierungsjahre)
].copy()

reentry_pooled_df = (
    reentry_calibration_base
    .groupby(
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "Altersklasse"
        ],
        as_index=False
    )
    .agg(
        summe_reentries=("summe_reentries", "sum"),
        summe_exits=("summe_exits", "sum"),
        anzahl_betrachtete_jahre=("Berichtsjahr", "nunique")
    )
)

reentry_pooled_df["pooled_reentry_quote_exits"] = np.where(
    reentry_pooled_df["summe_exits"] > 0,
    reentry_pooled_df["summe_reentries"] / reentry_pooled_df["summe_exits"],
    0
)

reentry_pooled_df["pooled_reentry_quote_exits"] = (
    reentry_pooled_df["pooled_reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_pooled_df = reentry_pooled_df.sort_values(
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start"
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 12. Kontrolle anzeigen
# ------------------------------------------------------------

print(f"Kalibrierungsfenster: {startjahr}-{aktuelles_jahr}")
print(f"Anzahl betrachtete Jahre: {anzahl_betrachtete_jahre}")
print(f"Breite der Altersklassen: {altersklassen_breite} Jahre")
print(f"Maximale Reentry-Quote: {max_reentry_quote}")

print("\nErster Output: reentry_yearly_df")
display(reentry_yearly_df.head())

print("\nZweiter Output: reentry_pooled_df")
display(reentry_pooled_df.head())
```

# Bestandsfortschreibung von 2025 auf 2026

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

basisjahr = 2025
prognosejahr = 2026

altersklassen_breite = 5

# Maximale Wahrscheinlichkeiten zur Sicherheit begrenzen
max_hazard_rate = 1.0
max_reentry_quote = 0.9

# WICHTIG:
# True, wenn bestand_df["Berichtsjahr"] bereits vorher so angepasst wurde,
# dass z.B. aus 01.01.2026 das Jahr 2025 wurde.
bestand_berichtsjahr_ist_bereits_angepasst = True

# Benötigte Input-DataFrames:
# bestand_df          = Bestand
# nzl_df              = Neuzulassungen
# hazard_pooled_df    = gepoolte Ausscheidewahrscheinlichkeiten
# reentry_pooled_df   = gepoolte Reentry-Quoten
# fleet_total_df      = Ziel-Gesamtbestand je Jahr


# ------------------------------------------------------------
# Hilfsfunktionen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr", bestand_logik=False):
    """
    Wandelt Berichtsjahr robust in ein Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Wenn bestand_logik=True:
    Ein Bestand mit Datum 01.01.2026 wird als Bestand des Jahres 2025 interpretiert.
    Deshalb wird dann minus 1 gerechnet.

    Wenn bestand_df["Berichtsjahr"] bereits vorher angepasst wurde,
    muss bestand_logik=False verwendet werden.
    """
    s = df[spalte]

    jahr_num = pd.to_numeric(s, errors="coerce")
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    jahr = jahr_num.fillna(jahr_date)

    if bestand_logik:
        jahr = jahr - 1

    return jahr.astype("Int64")


def berechne_altersklasse(df, alter_spalte="Alter", breite=5):
    """
    Erstellt Altersklasse_Start, Altersklasse_Ende und Altersklasse.
    """
    df["Altersklasse_Start"] = (df[alter_spalte] // breite) * breite
    df["Altersklasse_Ende"] = df["Altersklasse_Start"] + breite - 1

    df["Altersklasse"] = (
        df["Altersklasse_Start"].astype(int).astype(str)
        + "-"
        + df["Altersklasse_Ende"].astype(int).astype(str)
    )

    return df


# ------------------------------------------------------------
# 1. Bestand 2025 bereinigen und aggregieren
# ------------------------------------------------------------

bestand_work = bestand_df.copy()

bestand_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung",
    "Anzahl"
]

bestand_work = bestand_work[bestand_relevante_spalten].copy()

# KORRIGIERTE VERSION:
# Wenn Berichtsjahr im Bestand bereits angepasst ist, wird NICHT erneut -1 gerechnet.
bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
    bestand_work,
    "Berichtsjahr",
    bestand_logik=not bestand_berichtsjahr_ist_bereits_angepasst
)

bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
    bestand_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")

bestand_work["Anzahl"] = pd.to_numeric(
    bestand_work["Anzahl"],
    errors="coerce"
)

bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

bestand_2025 = bestand_work[
    bestand_work["Berichtsjahr"] == basisjahr
].copy()

print("Gefundene Zeilen für Bestand 2025:", len(bestand_2025))
print("Summe Bestand 2025:", bestand_2025["Anzahl"].sum())

if bestand_2025.empty:
    raise ValueError(
        f"Für das Basisjahr {basisjahr} wurde kein Bestand gefunden. "
        "Prüfe, ob bestand_df['Berichtsjahr'] bereits angepasst wurde "
        "und ob die Einstellung bestand_berichtsjahr_ist_bereits_angepasst korrekt ist."
    )

bestand_2025 = (
    bestand_2025
    .groupby(
        ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
        as_index=False
    )["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand_2025"})
)

# Alter im Basisjahr berechnen
bestand_2025["Alter"] = (
    basisjahr - bestand_2025["Jahr der Erstzulassung"]
)

bestand_2025 = bestand_2025[
    bestand_2025["Alter"] >= 0
].copy()

bestand_2025 = berechne_altersklasse(
    bestand_2025,
    alter_spalte="Alter",
    breite=altersklassen_breite
)


# ------------------------------------------------------------
# 2. Ausscheidewahrscheinlichkeiten vorbereiten
# ------------------------------------------------------------

hazard_rates = hazard_pooled_df.copy()

hazard_rates["pooled_hazard_rate"] = pd.to_numeric(
    hazard_rates["pooled_hazard_rate"],
    errors="coerce"
).fillna(0)

hazard_rates["pooled_hazard_rate"] = (
    hazard_rates["pooled_hazard_rate"]
    .clip(lower=0, upper=max_hazard_rate)
)

hazard_rates = hazard_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_hazard_rate"
    ]
].copy()


# ------------------------------------------------------------
# 3. Reentry-Quoten vorbereiten
# ------------------------------------------------------------

reentry_rates = reentry_pooled_df.copy()

reentry_rates["pooled_reentry_quote_exits"] = pd.to_numeric(
    reentry_rates["pooled_reentry_quote_exits"],
    errors="coerce"
).fillna(0)

reentry_rates["pooled_reentry_quote_exits"] = (
    reentry_rates["pooled_reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_rates = reentry_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_reentry_quote_exits"
    ]
].copy()


# ------------------------------------------------------------
# 4. Bestand 2025 mit Hazard-Rate und Reentry-Quote verbinden
# ------------------------------------------------------------

projection_base = bestand_2025.merge(
    hazard_rates,
    on=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende"
    ],
    how="left"
)

projection_base = projection_base.merge(
    reentry_rates,
    on=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende"
    ],
    how="left"
)

# Fehlende Wahrscheinlichkeiten als 0 behandeln
projection_base["pooled_hazard_rate"] = (
    projection_base["pooled_hazard_rate"]
    .fillna(0)
    .clip(lower=0, upper=max_hazard_rate)
)

projection_base["pooled_reentry_quote_exits"] = (
    projection_base["pooled_reentry_quote_exits"]
    .fillna(0)
    .clip(lower=0, upper=max_reentry_quote)
)


# ------------------------------------------------------------
# 5. Exits 2026 und Reentries 2026 berechnen
# ------------------------------------------------------------

projection_base["exits_2026"] = (
    projection_base["bestand_2025"]
    * projection_base["pooled_hazard_rate"]
)

projection_base["reentries_2026"] = (
    projection_base["exits_2026"]
    * projection_base["pooled_reentry_quote_exits"]
)

projection_base["bestand_nach_exits_reentries_2026"] = (
    projection_base["bestand_2025"]
    - projection_base["exits_2026"]
    + projection_base["reentries_2026"]
)

projection_base["bestand_nach_exits_reentries_2026"] = (
    projection_base["bestand_nach_exits_reentries_2026"]
    .clip(lower=0)
)


# ------------------------------------------------------------
# 6. Ziel-Gesamtbestand 2026 aus fleet_total_df holen
# ------------------------------------------------------------

fleet_total_work = fleet_total_df.copy()

fleet_total_work["Jahr"] = pd.to_numeric(
    fleet_total_work["Jahr"],
    errors="coerce"
).astype("Int64")

fleet_total_work["gesamtbestand_target"] = pd.to_numeric(
    fleet_total_work["gesamtbestand_target"],
    errors="coerce"
)

zielbestand_2026_series = fleet_total_work.loc[
    fleet_total_work["Jahr"] == prognosejahr,
    "gesamtbestand_target"
]

if zielbestand_2026_series.empty:
    raise ValueError(
        f"Für das Prognosejahr {prognosejahr} wurde kein Zielbestand "
        "in fleet_total_df gefunden."
    )

zielbestand_2026 = zielbestand_2026_series.iloc[0]

bestand_nach_exits_reentries = (
    projection_base["bestand_nach_exits_reentries_2026"].sum()
)

fehlende_fahrzeuge = max(
    zielbestand_2026 - bestand_nach_exits_reentries,
    0
)


# ------------------------------------------------------------
# 7. Verteilung der neuen Fahrzeuge nach NZL-Verteilung 2025
# ------------------------------------------------------------

nzl_work = nzl_df.copy()

nzl_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Anzahl"
]

nzl_work = nzl_work[nzl_relevante_spalten].copy()

nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(
    nzl_work,
    "Berichtsjahr",
    bestand_logik=False
)

nzl_work["Anzahl"] = pd.to_numeric(
    nzl_work["Anzahl"],
    errors="coerce"
)

nzl_work = nzl_work.dropna(subset=nzl_relevante_spalten)

nzl_2025 = nzl_work[
    nzl_work["Berichtsjahr"] == basisjahr
].copy()

nzl_verteilung = (
    nzl_2025
    .groupby(["Segment", "Antriebsart"], as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "nzl_2025"})
)

summe_nzl_2025 = nzl_verteilung["nzl_2025"].sum()

if summe_nzl_2025 > 0:
    nzl_verteilung["anteil_nzl_2025"] = (
        nzl_verteilung["nzl_2025"] / summe_nzl_2025
    )
else:
    raise ValueError(
        f"Für das Jahr {basisjahr} wurden keine Neuzulassungen gefunden. "
        "Die Verteilung neuer Fahrzeuge kann daher nicht berechnet werden."
    )

nzl_verteilung["neuzulassungen_2026"] = (
    fehlende_fahrzeuge * nzl_verteilung["anteil_nzl_2025"]
)

# Neue Fahrzeuge des Jahres 2026 werden als Kohorte Erstzulassungsjahr 2026 angelegt
new_vehicles_2026 = nzl_verteilung[
    ["Segment", "Antriebsart", "neuzulassungen_2026"]
].copy()

new_vehicles_2026["Jahr der Erstzulassung"] = prognosejahr

new_vehicles_2026 = new_vehicles_2026.rename(
    columns={"neuzulassungen_2026": "bestand_2026"}
)


# ------------------------------------------------------------
# 8. Fortgeschriebenen Bestand 2026 aus Altbestand + neuen Fahrzeugen erstellen
# ------------------------------------------------------------

bestand_alt_2026 = projection_base[
    [
        "Segment",
        "Antriebsart",
        "Jahr der Erstzulassung",
        "bestand_nach_exits_reentries_2026"
    ]
].copy()

bestand_alt_2026 = bestand_alt_2026.rename(
    columns={"bestand_nach_exits_reentries_2026": "bestand_2026"}
)

bestand_2026_df = pd.concat(
    [
        bestand_alt_2026,
        new_vehicles_2026
    ],
    ignore_index=True
)

bestand_2026_df = (
    bestand_2026_df
    .groupby(
        ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
        as_index=False
    )["bestand_2026"]
    .sum()
)

bestand_2026_df["Berichtsjahr"] = prognosejahr

bestand_2026_df = bestand_2026_df[
    [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Jahr der Erstzulassung",
        "bestand_2026"
    ]
].copy()


# ------------------------------------------------------------
# 9. Kontrolle ausgeben
# ------------------------------------------------------------

finaler_bestand_2026 = bestand_2026_df["bestand_2026"].sum()

print(f"Zielbestand {prognosejahr}: {zielbestand_2026:,.0f}")
print(
    f"Bestand nach Abgängen und Rückkehrern {prognosejahr}: "
    f"{bestand_nach_exits_reentries:,.0f}"
)
print(f"Fehlende Fahrzeuge / Neuzulassungen {prognosejahr}: {fehlende_fahrzeuge:,.0f}")
print(f"Finaler Bestand {prognosejahr}: {finaler_bestand_2026:,.0f}")

print("\nSumme erwartete Exits 2026:")
print(f"{projection_base['exits_2026'].sum():,.0f}")

print("\nSumme erwartete Reentries 2026:")
print(f"{projection_base['reentries_2026'].sum():,.0f}")

print("\nFortgeschriebener Bestand 2026:")
display(bestand_2026_df.head())

print("\nKontrolltabelle je Kohorte:")
display(projection_base.head())
```

# Fortschreibung bis 2050

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Annahmen / Einstellungen
# ------------------------------------------------------------

basisjahr = 2025
zieljahr = 2050

altersklassen_breite = 5

# Wahrscheinlichkeiten sinnvoll begrenzen
max_hazard_rate = 1.0
max_reentry_quote = 0.9

# WICHTIG:
# True, wenn bestand_df["Berichtsjahr"] bereits vorher so angepasst wurde,
# dass z.B. aus 01.01.2026 das Jahr 2025 wurde.
bestand_berichtsjahr_ist_bereits_angepasst = True

# Benötigte Input-DataFrames:
# bestand_df          = Bestand
# nzl_df              = Neuzulassungen
# hazard_pooled_df    = gepoolte Ausscheidewahrscheinlichkeiten
# reentry_pooled_df   = gepoolte Reentry-Quoten
# fleet_total_df      = Ziel-Gesamtbestand je Jahr


# ------------------------------------------------------------
# Hilfsfunktionen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr", bestand_logik=False):
    """
    Wandelt Berichtsjahr robust in ein Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Wenn bestand_logik=True:
    Ein Bestand mit Datum 01.01.2026 wird als Bestand des Jahres 2025 interpretiert.
    Deshalb wird dann minus 1 gerechnet.
    """
    s = df[spalte]

    jahr_num = pd.to_numeric(s, errors="coerce")
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    jahr = jahr_num.fillna(jahr_date)

    if bestand_logik:
        jahr = jahr - 1

    return jahr.astype("Int64")


def berechne_altersklasse(df, aktuelles_jahr, breite=5):
    """
    Berechnet Alter und 5-Jahres-Altersklasse für den aktuellen Bestand.
    """
    df["Alter"] = aktuelles_jahr - df["Jahr der Erstzulassung"]

    df = df[df["Alter"] >= 0].copy()

    df["Altersklasse_Start"] = (df["Alter"] // breite) * breite
    df["Altersklasse_Ende"] = df["Altersklasse_Start"] + breite - 1

    df["Altersklasse"] = (
        df["Altersklasse_Start"].astype(int).astype(str)
        + "-"
        + df["Altersklasse_Ende"].astype(int).astype(str)
    )

    return df


def safe_sum(series):
    """
    Summiert robust und behandelt fehlende Werte als 0.
    """
    return pd.to_numeric(series, errors="coerce").fillna(0).sum()


# ------------------------------------------------------------
# 1. Ausscheidewahrscheinlichkeiten vorbereiten
# ------------------------------------------------------------

hazard_rates = hazard_pooled_df.copy()

hazard_rates["pooled_hazard_rate"] = pd.to_numeric(
    hazard_rates["pooled_hazard_rate"],
    errors="coerce"
).fillna(0)

hazard_rates["pooled_hazard_rate"] = (
    hazard_rates["pooled_hazard_rate"]
    .clip(lower=0, upper=max_hazard_rate)
)

hazard_rates = hazard_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_hazard_rate"
    ]
].copy()

hazard_rates["Altersklasse_Start"] = pd.to_numeric(
    hazard_rates["Altersklasse_Start"],
    errors="coerce"
).astype("Int64")

hazard_rates["Altersklasse_Ende"] = pd.to_numeric(
    hazard_rates["Altersklasse_Ende"],
    errors="coerce"
).astype("Int64")

hazard_rates = hazard_rates.dropna(
    subset=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_hazard_rate"
    ]
)


# ------------------------------------------------------------
# 2. Rückkehrquoten vorbereiten
# ------------------------------------------------------------

reentry_rates = reentry_pooled_df.copy()

reentry_rates["pooled_reentry_quote_exits"] = pd.to_numeric(
    reentry_rates["pooled_reentry_quote_exits"],
    errors="coerce"
).fillna(0)

reentry_rates["pooled_reentry_quote_exits"] = (
    reentry_rates["pooled_reentry_quote_exits"]
    .clip(lower=0, upper=max_reentry_quote)
)

reentry_rates = reentry_rates[
    [
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_reentry_quote_exits"
    ]
].copy()

reentry_rates["Altersklasse_Start"] = pd.to_numeric(
    reentry_rates["Altersklasse_Start"],
    errors="coerce"
).astype("Int64")

reentry_rates["Altersklasse_Ende"] = pd.to_numeric(
    reentry_rates["Altersklasse_Ende"],
    errors="coerce"
).astype("Int64")

reentry_rates = reentry_rates.dropna(
    subset=[
        "Segment",
        "Antriebsart",
        "Altersklasse_Start",
        "Altersklasse_Ende",
        "pooled_reentry_quote_exits"
    ]
)


# ------------------------------------------------------------
# 3. Feste Neuzulassungsverteilung aus 2025 berechnen
# ------------------------------------------------------------

nzl_work = nzl_df.copy()

nzl_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Anzahl"
]

nzl_work = nzl_work[nzl_relevante_spalten].copy()

nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(
    nzl_work,
    "Berichtsjahr",
    bestand_logik=False
)

nzl_work["Anzahl"] = pd.to_numeric(
    nzl_work["Anzahl"],
    errors="coerce"
)

nzl_work = nzl_work.dropna(subset=nzl_relevante_spalten)

nzl_2025 = nzl_work[
    nzl_work["Berichtsjahr"] == basisjahr
].copy()

nzl_verteilung = (
    nzl_2025
    .groupby(["Segment", "Antriebsart"], as_index=False)["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "nzl_2025"})
)

summe_nzl_2025 = safe_sum(nzl_verteilung["nzl_2025"])

if summe_nzl_2025 <= 0:
    raise ValueError(
        f"Für das Jahr {basisjahr} wurden keine Neuzulassungen gefunden. "
        "Die Verteilung neuer Fahrzeuge kann daher nicht berechnet werden."
    )

nzl_verteilung["anteil_nzl_2025"] = (
    nzl_verteilung["nzl_2025"] / summe_nzl_2025
)

nzl_verteilung = nzl_verteilung[
    ["Segment", "Antriebsart", "anteil_nzl_2025"]
].copy()


# ------------------------------------------------------------
# 4. Ziel-Gesamtbestände vorbereiten
# ------------------------------------------------------------

fleet_total_work = fleet_total_df.copy()

fleet_total_work["Jahr"] = pd.to_numeric(
    fleet_total_work["Jahr"],
    errors="coerce"
).astype("Int64")

fleet_total_work["gesamtbestand_target"] = pd.to_numeric(
    fleet_total_work["gesamtbestand_target"],
    errors="coerce"
)

fleet_total_work = fleet_total_work.dropna(
    subset=["Jahr", "gesamtbestand_target"]
)

target_dict = dict(
    zip(
        fleet_total_work["Jahr"].astype(int),
        fleet_total_work["gesamtbestand_target"]
    )
)

fehlende_zieljahre = [
    jahr for jahr in range(basisjahr, zieljahr + 1)
    if jahr not in target_dict
]

if fehlende_zieljahre:
    raise ValueError(
        "Für folgende Jahre fehlen Zielbestände in fleet_total_df: "
        + ", ".join(map(str, fehlende_zieljahre))
    )


# ------------------------------------------------------------
# 5. Startbestand 2025 vorbereiten
# ------------------------------------------------------------

bestand_work = bestand_df.copy()

bestand_relevante_spalten = [
    "Berichtsjahr",
    "Segment",
    "Antriebsart",
    "Jahr der Erstzulassung",
    "Anzahl"
]

bestand_work = bestand_work[bestand_relevante_spalten].copy()

bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
    bestand_work,
    "Berichtsjahr",
    bestand_logik=not bestand_berichtsjahr_ist_bereits_angepasst
)

bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
    bestand_work["Jahr der Erstzulassung"],
    errors="coerce"
).astype("Int64")

bestand_work["Anzahl"] = pd.to_numeric(
    bestand_work["Anzahl"],
    errors="coerce"
)

bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

current_stock = bestand_work[
    bestand_work["Berichtsjahr"] == basisjahr
].copy()

if current_stock.empty:
    raise ValueError(
        f"Für das Basisjahr {basisjahr} wurde kein Bestand gefunden. "
        "Prüfe die Spalte Berichtsjahr und die Einstellung "
        "bestand_berichtsjahr_ist_bereits_angepasst."
    )

current_stock = (
    current_stock
    .groupby(
        ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
        as_index=False
    )["Anzahl"]
    .sum()
    .rename(columns={"Anzahl": "bestand"})
)

current_stock["Jahr der Erstzulassung"] = current_stock[
    "Jahr der Erstzulassung"
].astype(int)

current_stock["bestand"] = pd.to_numeric(
    current_stock["bestand"],
    errors="coerce"
).fillna(0)

current_stock = current_stock[current_stock["bestand"] > 0].copy()


# ------------------------------------------------------------
# 6. Mehrjährige Fortschreibung 2025 bis 2050
# ------------------------------------------------------------

results_by_year = {}

overview_rows = []

# Optional: Startjahr separat speichern
results_by_year[basisjahr] = {
    "finaler_bestand": current_stock.assign(Berichtsjahr=basisjahr).copy()
}

overview_rows.append({
    "Jahr": basisjahr,
    "zielbestand": target_dict[basisjahr],
    "bestand_vor_neuen_fahrzeugen": safe_sum(current_stock["bestand"]),
    "fehlende_fahrzeuge": max(
        target_dict[basisjahr] - safe_sum(current_stock["bestand"]),
        0
    ),
    "finaler_bestand": safe_sum(current_stock["bestand"])
})


for jahr in range(basisjahr, zieljahr):

    folgejahr = jahr + 1

    # --------------------------------------------------------
    # 6.1 Alter und Altersklassen im aktuellen Jahr berechnen
    # --------------------------------------------------------

    projection_base = current_stock.copy()

    projection_base = berechne_altersklasse(
        projection_base,
        aktuelles_jahr=jahr,
        breite=altersklassen_breite
    )

    # --------------------------------------------------------
    # 6.2 Hazard-Raten und Reentry-Quoten ergänzen
    # --------------------------------------------------------

    projection_base = projection_base.merge(
        hazard_rates,
        on=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende"
        ],
        how="left"
    )

    projection_base = projection_base.merge(
        reentry_rates,
        on=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende"
        ],
        how="left"
    )

    projection_base["pooled_hazard_rate"] = (
        projection_base["pooled_hazard_rate"]
        .fillna(0)
        .clip(lower=0, upper=max_hazard_rate)
    )

    projection_base["pooled_reentry_quote_exits"] = (
        projection_base["pooled_reentry_quote_exits"]
        .fillna(0)
        .clip(lower=0, upper=max_reentry_quote)
    )

    # --------------------------------------------------------
    # 6.3 Exits und Reentries für das Folgejahr berechnen
    # --------------------------------------------------------

    projection_base[f"exits_{folgejahr}"] = (
        projection_base["bestand"]
        * projection_base["pooled_hazard_rate"]
    )

    projection_base[f"reentries_{folgejahr}"] = (
        projection_base[f"exits_{folgejahr}"]
        * projection_base["pooled_reentry_quote_exits"]
    )

    projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"] = (
        projection_base["bestand"]
        - projection_base[f"exits_{folgejahr}"]
        + projection_base[f"reentries_{folgejahr}"]
    )

    projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"] = (
        projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"]
        .clip(lower=0)
    )

    bestehender_bestand_folgejahr = safe_sum(
        projection_base[f"bestand_vor_neuen_fahrzeugen_{folgejahr}"]
    )

    zielbestand_folgejahr = target_dict[folgejahr]

    fehlende_fahrzeuge = max(
        zielbestand_folgejahr - bestehender_bestand_folgejahr,
        0
    )

    # --------------------------------------------------------
    # 6.4 Neue Fahrzeuge auf Basis der NZL-Verteilung 2025 ergänzen
    # --------------------------------------------------------

    new_vehicles = nzl_verteilung.copy()

    new_vehicles["Jahr der Erstzulassung"] = folgejahr

    new_vehicles["neue_fahrzeuge"] = (
        fehlende_fahrzeuge * new_vehicles["anteil_nzl_2025"]
    )

    new_vehicles = new_vehicles[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            "neue_fahrzeuge"
        ]
    ].copy()

    # --------------------------------------------------------
    # 6.5 Finalen Bestand des Folgejahres erstellen
    # --------------------------------------------------------

    stock_existing_next = projection_base[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            f"bestand_vor_neuen_fahrzeugen_{folgejahr}"
        ]
    ].copy()

    stock_existing_next = stock_existing_next.rename(
        columns={
            f"bestand_vor_neuen_fahrzeugen_{folgejahr}": "bestand"
        }
    )

    stock_new_next = new_vehicles.rename(
        columns={"neue_fahrzeuge": "bestand"}
    )

    current_stock_next = pd.concat(
        [
            stock_existing_next,
            stock_new_next
        ],
        ignore_index=True
    )

    current_stock_next = (
        current_stock_next
        .groupby(
            ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
            as_index=False
        )["bestand"]
        .sum()
    )

    current_stock_next["bestand"] = pd.to_numeric(
        current_stock_next["bestand"],
        errors="coerce"
    ).fillna(0)

    current_stock_next = current_stock_next[
        current_stock_next["bestand"] > 0
    ].copy()

    finaler_bestand_folgejahr = safe_sum(current_stock_next["bestand"])

    # --------------------------------------------------------
    # 6.6 Ergebnisse für dieses Jahr speichern
    # --------------------------------------------------------

    exits_df = projection_base[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            f"exits_{folgejahr}"
        ]
    ].copy().rename(
        columns={f"exits_{folgejahr}": "exits"}
    )

    exits_df["Berichtsjahr"] = folgejahr

    reentries_df = projection_base[
        [
            "Segment",
            "Antriebsart",
            "Jahr der Erstzulassung",
            f"reentries_{folgejahr}"
        ]
    ].copy().rename(
        columns={f"reentries_{folgejahr}": "reentries"}
    )

    reentries_df["Berichtsjahr"] = folgejahr

    bestand_vor_neu_df = stock_existing_next.copy()
    bestand_vor_neu_df["Berichtsjahr"] = folgejahr
    bestand_vor_neu_df = bestand_vor_neu_df.rename(
        columns={"bestand": "bestand_vor_neuen_fahrzeugen"}
    )

    neue_fahrzeuge_df = new_vehicles.copy()
    neue_fahrzeuge_df["Berichtsjahr"] = folgejahr

    finaler_bestand_df = current_stock_next.copy()
    finaler_bestand_df["Berichtsjahr"] = folgejahr
    finaler_bestand_df = finaler_bestand_df.rename(
        columns={"bestand": "finaler_bestand"}
    )

    results_by_year[folgejahr] = {
        "abgaenge": exits_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "exits"
            ]
        ].copy(),

        "rueckkehrer": reentries_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "reentries"
            ]
        ].copy(),

        "bestand_vor_neuen_fahrzeugen": bestand_vor_neu_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "bestand_vor_neuen_fahrzeugen"
            ]
        ].copy(),

        "fehlende_fahrzeuge": pd.DataFrame({
            "Berichtsjahr": [folgejahr],
            "fehlende_fahrzeuge": [fehlende_fahrzeuge]
        }),

        "neue_fahrzeuge": neue_fahrzeuge_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "neue_fahrzeuge"
            ]
        ].copy(),

        "finaler_bestand": finaler_bestand_df[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "finaler_bestand"
            ]
        ].copy()
    }

    overview_rows.append({
        "Jahr": folgejahr,
        "zielbestand": zielbestand_folgejahr,
        "bestand_vor_neuen_fahrzeugen": bestehender_bestand_folgejahr,
        "fehlende_fahrzeuge": fehlende_fahrzeuge,
        "finaler_bestand": finaler_bestand_folgejahr
    })

    # Finaler Bestand des Folgejahres wird Startbestand für das nächste Jahr
    current_stock = current_stock_next.copy()


# ------------------------------------------------------------
# 7. Kompakte Übersichtstabelle erstellen
# ------------------------------------------------------------

fleet_projection_overview_df = pd.DataFrame(overview_rows)

fleet_projection_overview_df["abweichung_final_vs_ziel"] = (
    fleet_projection_overview_df["finaler_bestand"]
    - fleet_projection_overview_df["zielbestand"]
)

fleet_projection_overview_df = fleet_projection_overview_df[
    [
        "Jahr",
        "zielbestand",
        "bestand_vor_neuen_fahrzeugen",
        "fehlende_fahrzeuge",
        "finaler_bestand",
        "abweichung_final_vs_ziel"
    ]
].copy()


# ------------------------------------------------------------
# 8. Kontrolle anzeigen
# ------------------------------------------------------------

print(f"Fortschreibung von {basisjahr} bis {zieljahr}")
print(f"Startbestand {basisjahr}: {safe_sum(results_by_year[basisjahr]['finaler_bestand']['bestand']):,.0f}")
print(f"Zielbestand {zieljahr}: {target_dict[zieljahr]:,.0f}")
print(f"Finaler Bestand {zieljahr}: {fleet_projection_overview_df.loc[fleet_projection_overview_df['Jahr'] == zieljahr, 'finaler_bestand'].iloc[0]:,.0f}")

print("\nÜbersicht:")
display(fleet_projection_overview_df.head())
display(fleet_projection_overview_df.tail())

print("\nBeispiel: Ergebnisse für 2026")
display(results_by_year[2026]["finaler_bestand"].head())
display(results_by_year[2026]["abgaenge"].head())
display(results_by_year[2026]["rueckkehrer"].head())
display(results_by_year[2026]["neue_fahrzeuge"].head())
```

```python
# =====================================================
# PARAMETER BLOCK
# =====================================================
PARAMS = {
    # Zeitachse
    "START_YEAR": 2025,
    "END_YEAR_MULTI": 2050,
    "BIN_WIDTH": 5,

    # Modellgrenzen
    "MAX_REENTRY_RATE": 0.9,
    "MAX_HAZARD_RATE": 1.0,

    # Policy: Verbrenneraus
    "ICE_PHASEOUT_START_YEAR": 2026,
    "ICE_PHASEOUT_YEAR": 2035,

    # Szenarien
    "SCENARIOS": ["baseline", "verbrenneraus"],

    # Klassifikation: welche Antriebe als ICE gelten (lowercase match)
    "ICE_PATTERNS": ["benzin", "diesel", "hybrid", "plug", "phev", "verbren"],
    "NON_ICE_PATTERNS": ["elekt", "bev", "fcev", "wasserstoff"],
}
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Voraussetzung:
# PARAMS ist bereits definiert.
#
# Benötigte Input-DataFrames:
# bestand_df          = Bestand
# nzl_df              = Neuzulassungen
# hazard_pooled_df    = gepoolte Ausscheidewahrscheinlichkeiten
# reentry_pooled_df   = gepoolte Reentry-Quoten
# fleet_total_df      = Ziel-Gesamtbestand je Jahr
# ------------------------------------------------------------


# ------------------------------------------------------------
# Hilfsfunktionen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr", bestand_logik=False):
    """
    Wandelt Berichtsjahr robust in ein Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Wenn bestand_logik=True:
    Ein Bestand mit Datum 01.01.2026 wird als Bestand des Jahres 2025 interpretiert.
    Dann wird minus 1 gerechnet.
    """
    s = df[spalte]

    jahr_num = pd.to_numeric(s, errors="coerce")
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    jahr = jahr_num.fillna(jahr_date)

    if bestand_logik:
        jahr = jahr - 1

    return jahr.astype("Int64")


def safe_sum(series):
    """
    Summiert robust und behandelt fehlende Werte als 0.
    """
    return pd.to_numeric(series, errors="coerce").fillna(0).sum()


def berechne_altersklasse(df, aktuelles_jahr, breite):
    """
    Berechnet Alter und Altersklasse für den aktuellen Bestand.
    """
    df = df.copy()

    df["Alter"] = aktuelles_jahr - df["Jahr der Erstzulassung"]
    df = df[df["Alter"] >= 0].copy()

    df["Altersklasse_Start"] = (df["Alter"] // breite) * breite
    df["Altersklasse_Ende"] = df["Altersklasse_Start"] + breite - 1

    df["Altersklasse"] = (
        df["Altersklasse_Start"].astype(int).astype(str)
        + "-"
        + df["Altersklasse_Ende"].astype(int).astype(str)
    )

    return df


def ist_verbrenner(antriebsart, params=PARAMS):
    """
    Erkennt, ob eine Antriebsart als Verbrenner gilt.

    Logik:
    - Vergleich in Kleinbuchstaben.
    - Enthält die Antriebsart ein Muster aus ICE_PATTERNS, gilt sie als Verbrenner.
    - Enthält sie gleichzeitig ein Muster aus NON_ICE_PATTERNS, gilt sie NICHT als Verbrenner.
    """
    text = str(antriebsart).lower()

    ice_match = any(pattern.lower() in text for pattern in params["ICE_PATTERNS"])
    non_ice_match = any(pattern.lower() in text for pattern in params["NON_ICE_PATTERNS"])

    if ice_match and not non_ice_match:
        return True
    else:
        return False


def passe_nzl_verteilung_an(base_distribution, jahr, scenario, params=PARAMS):
    """
    Passt die Neuzulassungsverteilung je nach Szenario an.

    baseline:
    - ursprüngliche Anteile bleiben unverändert.

    verbrenneraus:
    - Verbrenneranteile werden ab ICE_PHASEOUT_START_YEAR
      bis ICE_PHASEOUT_YEAR linear reduziert.
    - Ab ICE_PHASEOUT_YEAR werden Verbrenneranteile auf 0 gesetzt.
    - Danach wird normalisiert, sodass die Anteile wieder 1 ergeben.
    - Falls keine gültige Summe entsteht, wird auf die ursprüngliche
      Verteilung zurückgegriffen.
    """
    dist = base_distribution.copy()

    dist["anteil_original"] = pd.to_numeric(
        dist["anteil_original"],
        errors="coerce"
    ).fillna(0)

    if scenario == "baseline":
        dist["anteil_szenario"] = dist["anteil_original"]

    elif scenario == "verbrenneraus":
        start = params["ICE_PHASEOUT_START_YEAR"]
        end = params["ICE_PHASEOUT_YEAR"]

        if jahr < start:
            ice_factor = 1.0
        elif jahr >= end:
            ice_factor = 0.0
        else:
            # Lineare Reduktion:
            # im Startjahr noch fast voll, bis zum Endjahr auf 0
            ice_factor = (end - jahr) / (end - start)

        dist["anteil_szenario"] = np.where(
            dist["ist_verbrenner"],
            dist["anteil_original"] * ice_factor,
            dist["anteil_original"]
        )

        summe_anteile = safe_sum(dist["anteil_szenario"])

        if summe_anteile > 0:
            dist["anteil_szenario"] = dist["anteil_szenario"] / summe_anteile
        else:
            dist["anteil_szenario"] = dist["anteil_original"]

    else:
        raise ValueError(
            f"Unbekanntes Szenario: {scenario}. "
            f"Erlaubt sind: {params['SCENARIOS']}"
        )

    # Sicherheit: Falls nach der Anpassung numerische Probleme auftreten
    summe_final = safe_sum(dist["anteil_szenario"])

    if summe_final > 0:
        dist["anteil_szenario"] = dist["anteil_szenario"] / summe_final
    else:
        dist["anteil_szenario"] = dist["anteil_original"]

    return dist[
        [
            "Segment",
            "Antriebsart",
            "ist_verbrenner",
            "anteil_original",
            "anteil_szenario"
        ]
    ].copy()


def run_fleet_scenario(
    scenario,
    params,
    bestand_df,
    nzl_df,
    hazard_pooled_df,
    reentry_pooled_df,
    fleet_total_df,
    bestand_berichtsjahr_ist_bereits_angepasst=True
):
    """
    Führt ein komplettes Szenario für die Bestandsfortschreibung aus.

    Rückgabe:
    - scenario_results_by_year: Dictionary mit Abgängen, Rückkehrern,
      fehlenden Fahrzeugen, neuen Fahrzeugen und finalem Bestand je Jahr.
    - scenario_summary_df: kompakte Übersicht je Jahr.
    """

    start_year = params["START_YEAR"]
    end_year = params["END_YEAR_MULTI"]
    bin_width = params["BIN_WIDTH"]
    max_hazard_rate = params["MAX_HAZARD_RATE"]
    max_reentry_rate = params["MAX_REENTRY_RATE"]

    # --------------------------------------------------------
    # 1. Hazard-Raten vorbereiten
    # --------------------------------------------------------

    hazard_rates = hazard_pooled_df.copy()

    hazard_rates["pooled_hazard_rate"] = pd.to_numeric(
        hazard_rates["pooled_hazard_rate"],
        errors="coerce"
    ).fillna(0)

    hazard_rates["pooled_hazard_rate"] = (
        hazard_rates["pooled_hazard_rate"]
        .clip(lower=0, upper=max_hazard_rate)
    )

    hazard_rates = hazard_rates[
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_hazard_rate"
        ]
    ].copy()

    hazard_rates["Altersklasse_Start"] = pd.to_numeric(
        hazard_rates["Altersklasse_Start"],
        errors="coerce"
    ).astype("Int64")

    hazard_rates["Altersklasse_Ende"] = pd.to_numeric(
        hazard_rates["Altersklasse_Ende"],
        errors="coerce"
    ).astype("Int64")

    hazard_rates = hazard_rates.dropna(
        subset=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_hazard_rate"
        ]
    )

    # --------------------------------------------------------
    # 2. Reentry-Quoten vorbereiten
    # --------------------------------------------------------

    reentry_rates = reentry_pooled_df.copy()

    reentry_rates["pooled_reentry_quote_exits"] = pd.to_numeric(
        reentry_rates["pooled_reentry_quote_exits"],
        errors="coerce"
    ).fillna(0)

    reentry_rates["pooled_reentry_quote_exits"] = (
        reentry_rates["pooled_reentry_quote_exits"]
        .clip(lower=0, upper=max_reentry_rate)
    )

    reentry_rates = reentry_rates[
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_reentry_quote_exits"
        ]
    ].copy()

    reentry_rates["Altersklasse_Start"] = pd.to_numeric(
        reentry_rates["Altersklasse_Start"],
        errors="coerce"
    ).astype("Int64")

    reentry_rates["Altersklasse_Ende"] = pd.to_numeric(
        reentry_rates["Altersklasse_Ende"],
        errors="coerce"
    ).astype("Int64")

    reentry_rates = reentry_rates.dropna(
        subset=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_reentry_quote_exits"
        ]
    )

    # --------------------------------------------------------
    # 3. Basisverteilung der Neuzulassungen aus Startjahr
    # --------------------------------------------------------

    nzl_work = nzl_df.copy()

    nzl_relevante_spalten = [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Anzahl"
    ]

    nzl_work = nzl_work[nzl_relevante_spalten].copy()

    nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(
        nzl_work,
        "Berichtsjahr",
        bestand_logik=False
    )

    nzl_work["Anzahl"] = pd.to_numeric(
        nzl_work["Anzahl"],
        errors="coerce"
    )

    nzl_work = nzl_work.dropna(subset=nzl_relevante_spalten)

    nzl_start = nzl_work[
        nzl_work["Berichtsjahr"] == start_year
    ].copy()

    nzl_base_distribution = (
        nzl_start
        .groupby(["Segment", "Antriebsart"], as_index=False)["Anzahl"]
        .sum()
        .rename(columns={"Anzahl": "nzl_startjahr"})
    )

    summe_nzl_start = safe_sum(nzl_base_distribution["nzl_startjahr"])

    if summe_nzl_start <= 0:
        raise ValueError(
            f"Für das Startjahr {start_year} wurden keine gültigen "
            "Neuzulassungen gefunden. Die Verteilung neuer Fahrzeuge "
            "kann daher nicht berechnet werden."
        )

    nzl_base_distribution["anteil_original"] = (
        nzl_base_distribution["nzl_startjahr"] / summe_nzl_start
    )

    nzl_base_distribution["ist_verbrenner"] = (
        nzl_base_distribution["Antriebsart"].apply(
            lambda x: ist_verbrenner(x, params)
        )
    )

    nzl_base_distribution = nzl_base_distribution[
        [
            "Segment",
            "Antriebsart",
            "ist_verbrenner",
            "anteil_original"
        ]
    ].copy()

    # --------------------------------------------------------
    # 4. Ziel-Gesamtbestände vorbereiten
    # --------------------------------------------------------

    fleet_total_work = fleet_total_df.copy()

    fleet_total_work["Jahr"] = pd.to_numeric(
        fleet_total_work["Jahr"],
        errors="coerce"
    ).astype("Int64")

    fleet_total_work["gesamtbestand_target"] = pd.to_numeric(
        fleet_total_work["gesamtbestand_target"],
        errors="coerce"
    )

    fleet_total_work = fleet_total_work.dropna(
        subset=["Jahr", "gesamtbestand_target"]
    )

    target_dict = dict(
        zip(
            fleet_total_work["Jahr"].astype(int),
            fleet_total_work["gesamtbestand_target"]
        )
    )

    fehlende_zieljahre = [
        jahr for jahr in range(start_year, end_year + 1)
        if jahr not in target_dict
    ]

    if fehlende_zieljahre:
        raise ValueError(
            "Für folgende Jahre fehlen Zielbestände in fleet_total_df: "
            + ", ".join(map(str, fehlende_zieljahre))
        )

    # --------------------------------------------------------
    # 5. Startbestand vorbereiten
    # --------------------------------------------------------

    bestand_work = bestand_df.copy()

    bestand_relevante_spalten = [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Jahr der Erstzulassung",
        "Anzahl"
    ]

    bestand_work = bestand_work[bestand_relevante_spalten].copy()

    bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
        bestand_work,
        "Berichtsjahr",
        bestand_logik=not bestand_berichtsjahr_ist_bereits_angepasst
    )

    bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
        bestand_work["Jahr der Erstzulassung"],
        errors="coerce"
    ).astype("Int64")

    bestand_work["Anzahl"] = pd.to_numeric(
        bestand_work["Anzahl"],
        errors="coerce"
    )

    bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

    current_stock = bestand_work[
        bestand_work["Berichtsjahr"] == start_year
    ].copy()

    if current_stock.empty:
        raise ValueError(
            f"Für das Startjahr {start_year} wurde kein Bestand gefunden. "
            "Prüfe die Spalte Berichtsjahr und die Einstellung "
            "bestand_berichtsjahr_ist_bereits_angepasst."
        )

    current_stock = (
        current_stock
        .groupby(
            ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
            as_index=False
        )["Anzahl"]
        .sum()
        .rename(columns={"Anzahl": "bestand"})
    )

    current_stock["Jahr der Erstzulassung"] = pd.to_numeric(
        current_stock["Jahr der Erstzulassung"],
        errors="coerce"
    ).astype("Int64")

    current_stock["bestand"] = pd.to_numeric(
        current_stock["bestand"],
        errors="coerce"
    ).fillna(0)

    current_stock = current_stock.dropna(
        subset=["Segment", "Antriebsart", "Jahr der Erstzulassung"]
    )

    current_stock = current_stock[
        current_stock["bestand"] > 0
    ].copy()

    # --------------------------------------------------------
    # 6. Simulation Jahr für Jahr
    # --------------------------------------------------------

    scenario_results_by_year = {}
    summary_rows = []

    scenario_results_by_year[start_year] = {
        "abgaenge": pd.DataFrame(),
        "rueckkehrer": pd.DataFrame(),
        "fehlende_fahrzeuge": pd.DataFrame({
            "Berichtsjahr": [start_year],
            "fehlende_fahrzeuge": [
                max(target_dict[start_year] - safe_sum(current_stock["bestand"]), 0)
            ]
        }),
        "neue_fahrzeuge": pd.DataFrame(),
        "finaler_bestand": current_stock.assign(Berichtsjahr=start_year).rename(
            columns={"bestand": "finaler_bestand"}
        )[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "finaler_bestand"
            ]
        ].copy()
    }

    summary_rows.append({
        "Szenario": scenario,
        "Jahr": start_year,
        "zielbestand": target_dict[start_year],
        "bestand_vor_neuen_fahrzeugen": safe_sum(current_stock["bestand"]),
        "fehlende_fahrzeuge": max(
            target_dict[start_year] - safe_sum(current_stock["bestand"]),
            0
        ),
        "finaler_bestand": safe_sum(current_stock["bestand"])
    })

    for jahr in range(start_year, end_year):

        folgejahr = jahr + 1

        # ----------------------------------------------------
        # 6.1 Alter und Altersklassen berechnen
        # ----------------------------------------------------

        projection_base = current_stock.copy()

        projection_base = berechne_altersklasse(
            projection_base,
            aktuelles_jahr=jahr,
            breite=bin_width
        )

        # ----------------------------------------------------
        # 6.2 Hazard und Reentry ergänzen
        # ----------------------------------------------------

        projection_base = projection_base.merge(
            hazard_rates,
            on=[
                "Segment",
                "Antriebsart",
                "Altersklasse_Start",
                "Altersklasse_Ende"
            ],
            how="left"
        )

        projection_base = projection_base.merge(
            reentry_rates,
            on=[
                "Segment",
                "Antriebsart",
                "Altersklasse_Start",
                "Altersklasse_Ende"
            ],
            how="left"
        )

        projection_base["pooled_hazard_rate"] = (
            projection_base["pooled_hazard_rate"]
            .fillna(0)
            .clip(lower=0, upper=max_hazard_rate)
        )

        projection_base["pooled_reentry_quote_exits"] = (
            projection_base["pooled_reentry_quote_exits"]
            .fillna(0)
            .clip(lower=0, upper=max_reentry_rate)
        )

        # ----------------------------------------------------
        # 6.3 Exits, Reentries und Bestand vor neuen Fahrzeugen
        # ----------------------------------------------------

        projection_base["exits"] = (
            projection_base["bestand"]
            * projection_base["pooled_hazard_rate"]
        )

        projection_base["reentries"] = (
            projection_base["exits"]
            * projection_base["pooled_reentry_quote_exits"]
        )

        projection_base["bestand_vor_neuen_fahrzeugen"] = (
            projection_base["bestand"]
            - projection_base["exits"]
            + projection_base["reentries"]
        ).clip(lower=0)

        bestehender_bestand = safe_sum(
            projection_base["bestand_vor_neuen_fahrzeugen"]
        )

        zielbestand = target_dict[folgejahr]

        fehlende_fahrzeuge = max(
            zielbestand - bestehender_bestand,
            0
        )

        # ----------------------------------------------------
        # 6.4 Szenariospezifische NZL-Verteilung anwenden
        # ----------------------------------------------------

        nzl_distribution_scenario = passe_nzl_verteilung_an(
            base_distribution=nzl_base_distribution,
            jahr=folgejahr,
            scenario=scenario,
            params=params
        )

        new_vehicles = nzl_distribution_scenario.copy()

        new_vehicles["Jahr der Erstzulassung"] = folgejahr
        new_vehicles["neue_fahrzeuge"] = (
            fehlende_fahrzeuge * new_vehicles["anteil_szenario"]
        )

        new_vehicles = new_vehicles[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "ist_verbrenner",
                "anteil_original",
                "anteil_szenario",
                "neue_fahrzeuge"
            ]
        ].copy()

        # ----------------------------------------------------
        # 6.5 Finalen Bestand des Folgejahres erstellen
        # ----------------------------------------------------

        stock_existing_next = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "bestand_vor_neuen_fahrzeugen"
            ]
        ].copy()

        stock_existing_next = stock_existing_next.rename(
            columns={"bestand_vor_neuen_fahrzeugen": "bestand"}
        )

        stock_new_next = new_vehicles[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "neue_fahrzeuge"
            ]
        ].copy()

        stock_new_next = stock_new_next.rename(
            columns={"neue_fahrzeuge": "bestand"}
        )

        current_stock_next = pd.concat(
            [
                stock_existing_next,
                stock_new_next
            ],
            ignore_index=True
        )

        current_stock_next["bestand"] = pd.to_numeric(
            current_stock_next["bestand"],
            errors="coerce"
        ).fillna(0)

        current_stock_next = (
            current_stock_next
            .groupby(
                ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
                as_index=False
            )["bestand"]
            .sum()
        )

        current_stock_next = current_stock_next[
            current_stock_next["bestand"] > 0
        ].copy()

        finaler_bestand = safe_sum(current_stock_next["bestand"])

        # ----------------------------------------------------
        # 6.6 Ergebnisse speichern
        # ----------------------------------------------------

        abgaenge_df = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "exits"
            ]
        ].copy()

        abgaenge_df["Berichtsjahr"] = folgejahr

        rueckkehrer_df = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "reentries"
            ]
        ].copy()

        rueckkehrer_df["Berichtsjahr"] = folgejahr

        bestand_vor_neu_df = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "bestand_vor_neuen_fahrzeugen"
            ]
        ].copy()

        bestand_vor_neu_df["Berichtsjahr"] = folgejahr

        fehlende_fahrzeuge_df = pd.DataFrame({
            "Berichtsjahr": [folgejahr],
            "fehlende_fahrzeuge": [fehlende_fahrzeuge]
        })

        finaler_bestand_df = current_stock_next.copy()
        finaler_bestand_df["Berichtsjahr"] = folgejahr
        finaler_bestand_df = finaler_bestand_df.rename(
            columns={"bestand": "finaler_bestand"}
        )

        scenario_results_by_year[folgejahr] = {
            "abgaenge": abgaenge_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "exits"
                ]
            ].copy(),

            "rueckkehrer": rueckkehrer_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "reentries"
                ]
            ].copy(),

            "bestand_vor_neuen_fahrzeugen": bestand_vor_neu_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "bestand_vor_neuen_fahrzeugen"
                ]
            ].copy(),

            "fehlende_fahrzeuge": fehlende_fahrzeuge_df.copy(),

            "neue_fahrzeuge": new_vehicles.assign(
                Berichtsjahr=folgejahr
            )[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "ist_verbrenner",
                    "anteil_original",
                    "anteil_szenario",
                    "neue_fahrzeuge"
                ]
            ].copy(),

            "finaler_bestand": finaler_bestand_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "finaler_bestand"
                ]
            ].copy()
        }

        summary_rows.append({
            "Szenario": scenario,
            "Jahr": folgejahr,
            "zielbestand": zielbestand,
            "bestand_vor_neuen_fahrzeugen": bestehender_bestand,
            "fehlende_fahrzeuge": fehlende_fahrzeuge,
            "finaler_bestand": finaler_bestand
        })

        # Finaler Bestand des Folgejahres wird Startbestand für nächste Iteration
        current_stock = current_stock_next.copy()

    # --------------------------------------------------------
    # 7. Summary-Tabelle
    # --------------------------------------------------------

    scenario_summary_df = pd.DataFrame(summary_rows)

    scenario_summary_df["abweichung_final_vs_ziel"] = (
        scenario_summary_df["finaler_bestand"]
        - scenario_summary_df["zielbestand"]
    )

    scenario_summary_df = scenario_summary_df[
        [
            "Szenario",
            "Jahr",
            "zielbestand",
            "bestand_vor_neuen_fahrzeugen",
            "fehlende_fahrzeuge",
            "finaler_bestand",
            "abweichung_final_vs_ziel"
        ]
    ].copy()

    return scenario_results_by_year, scenario_summary_df


# ------------------------------------------------------------
# Alle Szenarien automatisch durchrechnen
# ------------------------------------------------------------

scenario_outputs = {}

for scenario in PARAMS["SCENARIOS"]:

    scenario_results_by_year, scenario_summary_df = run_fleet_scenario(
        scenario=scenario,
        params=PARAMS,
        bestand_df=bestand_df,
        nzl_df=nzl_df,
        hazard_pooled_df=hazard_pooled_df,
        reentry_pooled_df=reentry_pooled_df,
        fleet_total_df=fleet_total_df,
        bestand_berichtsjahr_ist_bereits_angepasst=True
    )

    scenario_outputs[scenario] = {
        "results_by_year": scenario_results_by_year,
        "summary": scenario_summary_df
    }


# ------------------------------------------------------------
# Kontrolle ausgeben
# ------------------------------------------------------------

print("Berechnete Szenarien:")
print(list(scenario_outputs.keys()))

print("\nJahre im Basisszenario:")
print(list(scenario_outputs["baseline"]["results_by_year"].keys()))

print("\nSummary baseline:")
display(scenario_outputs["baseline"]["summary"].head())

print("\nSummary verbrenneraus:")
display(scenario_outputs["verbrenneraus"]["summary"].head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Hilfsfunktion: Pivot-Tabelle je Szenario erstellen
# ------------------------------------------------------------

def create_stock_by_drive_pivot(scenario_outputs, scenario_name):
    """
    Erstellt eine Pivot-Tabelle:
    Zeilen: Jahre
    Spalten: Antriebsarten
    Werte: Fahrzeugbestand
    """

    stock_by_drive_rows = []

    results_by_year = scenario_outputs[scenario_name]["results_by_year"]

    for jahr, year_results in results_by_year.items():

        stock_df = year_results["finaler_bestand"].copy()

        # Je nach Jahr kann die Bestandsspalte unterschiedlich heißen
        if "finaler_bestand" in stock_df.columns:
            bestand_spalte = "finaler_bestand"
        elif "bestand" in stock_df.columns:
            bestand_spalte = "bestand"
        else:
            raise ValueError(
                f"Im Szenario '{scenario_name}' wurde für das Jahr {jahr} "
                "keine Bestandsspalte gefunden. Erwartet wurde "
                "'finaler_bestand' oder 'bestand'."
            )

        stock_df[bestand_spalte] = pd.to_numeric(
            stock_df[bestand_spalte],
            errors="coerce"
        ).fillna(0)

        stock_by_drive = (
            stock_df
            .groupby("Antriebsart", as_index=False)[bestand_spalte]
            .sum()
            .rename(columns={bestand_spalte: "fahrzeugbestand"})
        )

        stock_by_drive["Jahr"] = jahr

        stock_by_drive_rows.append(stock_by_drive)

    stock_by_drive_year_df = pd.concat(
        stock_by_drive_rows,
        ignore_index=True
    )

    stock_by_drive_pivot = (
        stock_by_drive_year_df
        .pivot_table(
            index="Jahr",
            columns="Antriebsart",
            values="fahrzeugbestand",
            aggfunc="sum",
            fill_value=0
        )
        .sort_index()
    )

    return stock_by_drive_pivot


# ------------------------------------------------------------
# 2. Pivot-Tabellen für beide Szenarien erstellen
# ------------------------------------------------------------

baseline_pivot = create_stock_by_drive_pivot(
    scenario_outputs,
    "baseline"
)

verbrenneraus_pivot = create_stock_by_drive_pivot(
    scenario_outputs,
    "verbrenneraus"
)


# ------------------------------------------------------------
# 3. Spalten angleichen, damit beide Diagramme dieselben Antriebsarten haben
# ------------------------------------------------------------

all_drive_types = sorted(
    set(baseline_pivot.columns).union(set(verbrenneraus_pivot.columns))
)

baseline_pivot = baseline_pivot.reindex(
    columns=all_drive_types,
    fill_value=0
)

verbrenneraus_pivot = verbrenneraus_pivot.reindex(
    columns=all_drive_types,
    fill_value=0
)


# ------------------------------------------------------------
# 4. Gemeinsame y-Achse bestimmen
# ------------------------------------------------------------

y_max = max(
    baseline_pivot.sum(axis=1).max(),
    verbrenneraus_pivot.sum(axis=1).max()
)

y_max = y_max * 1.05


# ------------------------------------------------------------
# 5. Zwei gestapelte Balkendiagramme nebeneinander erstellen
# ------------------------------------------------------------

fig, axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(18, 7),
    sharey=True
)

baseline_pivot.plot(
    kind="bar",
    stacked=True,
    ax=axes[0],
    legend=False
)

axes[0].set_title("Baseline")
axes[0].set_xlabel("Year")
axes[0].set_ylabel("Vehicle stock")
axes[0].set_ylim(0, y_max)
axes[0].tick_params(axis="x", rotation=45)

verbrenneraus_pivot.plot(
    kind="bar",
    stacked=True,
    ax=axes[1],
    legend=False
)

axes[1].set_title("Verbrenneraus")
axes[1].set_xlabel("Year")
axes[1].set_ylabel("Vehicle stock")
axes[1].set_ylim(0, y_max)
axes[1].tick_params(axis="x", rotation=45)


# ------------------------------------------------------------
# 6. Gemeinsame Legende rechts außerhalb platzieren
# ------------------------------------------------------------

handles, labels = axes[1].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    title="Drive type",
    bbox_to_anchor=(1.02, 0.5),
    loc="center left"
)

plt.tight_layout(rect=[0, 0, 0.85, 1])
plt.show()


# ------------------------------------------------------------
# 7. Kontrolle: erste Zeilen der Pivot-Tabellen anzeigen
# ------------------------------------------------------------

print("Pivot-Tabelle baseline:")
display(baseline_pivot.head())

print("Pivot-Tabelle verbrenneraus:")
display(verbrenneraus_pivot.head())
```

# Erweiterte Policies: Diesel-Fahrverbot und Abwrackprämie

```python
import numpy as np
import pandas as pd

# =====================================================
# PARAMETER BLOCK (erweitert)
# =====================================================
PARAMS_EXT = {
    "START_YEAR": 2025,
    "END_YEAR_MULTI": 2050,
    "BIN_WIDTH": 5,
    "MAX_REENTRY_RATE": 0.9,
    "MAX_HAZARD_RATE": 1.0,

    # Verbrenneraus
    "ICE_PHASEOUT_START_YEAR": 2026,
    "ICE_PHASEOUT_YEAR": 2035,

    # Diesel-Fahrverbot: ab diesem Jahr dürfen Diesel nicht mehr im Bestand sein
    "DIESEL_BAN_YEAR": 2032,

    # Abwrackprämie: Hazard-Erhöhung über Multiplikator
    "SCRAPPAGE_HAZARD_MULTIPLIER": 3,
    "SCRAPPAGE_TARGET_PATTERNS": ["diesel", "benzin"],

    # Szenarien
    "SCENARIOS": ["baseline", "verbrenneraus", "diesel_fahrverbot", "abwrackpraemie"],

    # Textmuster für Antriebsklassifikation
    "ICE_PATTERNS": ["benzin", "diesel", "hybrid", "plug", "phev", "verbren"],
    "NON_ICE_PATTERNS": ["elekt", "bev", "fcev", "wasserstoff"],
    "DIESEL_PATTERNS": ["diesel"],
}
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Voraussetzung:
# PARAMS_EXT ist bereits definiert.
#
# Benötigte Input-DataFrames:
# bestand_df          = Bestand
# nzl_df              = Neuzulassungen
# hazard_pooled_df    = gepoolte Ausscheidewahrscheinlichkeiten
# reentry_pooled_df   = gepoolte Reentry-Quoten
# fleet_total_df      = Ziel-Gesamtbestand je Jahr
# ------------------------------------------------------------


# ------------------------------------------------------------
# Hilfsfunktionen
# ------------------------------------------------------------

def bereinige_berichtsjahr(df, spalte="Berichtsjahr", bestand_logik=False):
    """
    Wandelt Berichtsjahr robust in ein Jahr um.
    Funktioniert sowohl bei Jahreszahlen wie 2025
    als auch bei Datumswerten wie 01.01.2026.

    Wenn bestand_logik=True:
    Ein Bestand mit Datum 01.01.2026 wird als Bestand des Jahres 2025 interpretiert.
    Dann wird minus 1 gerechnet.
    """
    s = df[spalte]

    jahr_num = pd.to_numeric(s, errors="coerce")
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    jahr = jahr_num.fillna(jahr_date)

    if bestand_logik:
        jahr = jahr - 1

    return jahr.astype("Int64")


def safe_sum(series):
    """
    Summiert robust und behandelt fehlende Werte als 0.
    """
    return pd.to_numeric(series, errors="coerce").fillna(0).sum()


def contains_any_pattern(text, patterns):
    """
    Prüft, ob ein Text eines der angegebenen Muster enthält.
    Vergleich erfolgt in Kleinbuchstaben.
    """
    text_lower = str(text).lower()

    return any(
        str(pattern).lower() in text_lower
        for pattern in patterns
    )


def ist_verbrenner(antriebsart, params=PARAMS_EXT):
    """
    Erkennt, ob eine Antriebsart als Verbrenner gilt.

    Logik:
    - Enthält die Antriebsart ein Muster aus ICE_PATTERNS,
      wird sie grundsätzlich als Verbrenner erkannt.
    - Enthält sie gleichzeitig ein Muster aus NON_ICE_PATTERNS,
      wird sie NICHT als Verbrenner behandelt.
    """
    ice_match = contains_any_pattern(
        antriebsart,
        params["ICE_PATTERNS"]
    )

    non_ice_match = contains_any_pattern(
        antriebsart,
        params["NON_ICE_PATTERNS"]
    )

    return ice_match and not non_ice_match


def ist_diesel(antriebsart, params=PARAMS_EXT):
    """
    Erkennt Diesel-Antriebe anhand der DIESEL_PATTERNS.
    """
    return contains_any_pattern(
        antriebsart,
        params["DIESEL_PATTERNS"]
    )


def berechne_altersklasse(df, aktuelles_jahr, breite):
    """
    Berechnet Alter und Altersklasse für den aktuellen Bestand.
    """
    df = df.copy()

    df["Alter"] = aktuelles_jahr - df["Jahr der Erstzulassung"]
    df = df[df["Alter"] >= 0].copy()

    df["Altersklasse_Start"] = (df["Alter"] // breite) * breite
    df["Altersklasse_Ende"] = df["Altersklasse_Start"] + breite - 1

    df["Altersklasse"] = (
        df["Altersklasse_Start"].astype(int).astype(str)
        + "-"
        + df["Altersklasse_Ende"].astype(int).astype(str)
    )

    return df


def passe_nzl_verteilung_an(base_distribution, jahr, scenario, params=PARAMS_EXT):
    """
    Passt die Neuzulassungsverteilung je nach Szenario an.

    baseline:
    - ursprüngliche Anteile bleiben unverändert.

    verbrenneraus:
    - Verbrenneranteile werden ab ICE_PHASEOUT_START_YEAR
      bis ICE_PHASEOUT_YEAR linear reduziert.
    - Ab ICE_PHASEOUT_YEAR werden Verbrenneranteile auf 0 gesetzt.

    diesel_fahrverbot:
    - ab DIESEL_BAN_YEAR werden Dieselanteile bei neuen Fahrzeugen auf 0 gesetzt.

    abwrackpraemie:
    - Neuzulassungsverteilung bleibt unverändert.
      Der Policy-Effekt wird über erhöhte Hazard-Raten abgebildet.

    Nach jeder Anpassung werden die Anteile wieder auf Summe 1 normalisiert.
    Falls keine gültige Summe entsteht, wird auf die ursprüngliche Verteilung zurückgefallen.
    """
    dist = base_distribution.copy()

    dist["anteil_original"] = pd.to_numeric(
        dist["anteil_original"],
        errors="coerce"
    ).fillna(0)

    dist["anteil_szenario"] = dist["anteil_original"]

    if scenario == "baseline":
        pass

    elif scenario == "verbrenneraus":
        start = params["ICE_PHASEOUT_START_YEAR"]
        end = params["ICE_PHASEOUT_YEAR"]

        if jahr < start:
            ice_factor = 1.0
        elif jahr >= end:
            ice_factor = 0.0
        else:
            ice_factor = (end - jahr) / (end - start)

        dist["anteil_szenario"] = np.where(
            dist["ist_verbrenner"],
            dist["anteil_original"] * ice_factor,
            dist["anteil_original"]
        )

    elif scenario == "diesel_fahrverbot":
        ban_year = params["DIESEL_BAN_YEAR"]

        if jahr >= ban_year:
            dist["anteil_szenario"] = np.where(
                dist["ist_diesel"],
                0,
                dist["anteil_original"]
            )

    elif scenario == "abwrackpraemie":
        pass

    else:
        raise ValueError(
            f"Unbekanntes Szenario: {scenario}. "
            f"Erlaubt sind: {params['SCENARIOS']}"
        )

    summe_szenario = safe_sum(dist["anteil_szenario"])

    if summe_szenario > 0:
        dist["anteil_szenario"] = dist["anteil_szenario"] / summe_szenario
    else:
        dist["anteil_szenario"] = dist["anteil_original"]

    summe_final = safe_sum(dist["anteil_szenario"])

    if summe_final > 0:
        dist["anteil_szenario"] = dist["anteil_szenario"] / summe_final
    else:
        raise ValueError(
            f"Für Szenario '{scenario}' und Jahr {jahr} konnte keine gültige "
            "Neuzulassungsverteilung gebildet werden."
        )

    return dist[
        [
            "Segment",
            "Antriebsart",
            "ist_verbrenner",
            "ist_diesel",
            "anteil_original",
            "anteil_szenario"
        ]
    ].copy()


def run_policy_scenario(
    scenario,
    params,
    bestand_df,
    nzl_df,
    hazard_pooled_df,
    reentry_pooled_df,
    fleet_total_df,
    bestand_berichtsjahr_ist_bereits_angepasst=True
):
    """
    Berechnet ein einzelnes Politikszenario.

    Rückgabe:
    - scenario_results_by_year: Dictionary mit Abgängen, Rückkehrern,
      fehlenden Fahrzeugen, neuen Fahrzeugen und finalem Bestand je Jahr.
    - scenario_summary_df: kompakte Summary-Tabelle je Jahr.
    """

    start_year = params["START_YEAR"]
    end_year = params["END_YEAR_MULTI"]
    bin_width = params["BIN_WIDTH"]

    max_hazard_rate = params["MAX_HAZARD_RATE"]
    max_reentry_rate = params["MAX_REENTRY_RATE"]

    # --------------------------------------------------------
    # 1. Hazard-Raten vorbereiten
    # --------------------------------------------------------

    hazard_rates = hazard_pooled_df.copy()

    hazard_rates["pooled_hazard_rate"] = pd.to_numeric(
        hazard_rates["pooled_hazard_rate"],
        errors="coerce"
    ).fillna(0)

    hazard_rates["pooled_hazard_rate"] = (
        hazard_rates["pooled_hazard_rate"]
        .clip(lower=0, upper=max_hazard_rate)
    )

    hazard_rates = hazard_rates[
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_hazard_rate"
        ]
    ].copy()

    hazard_rates["Altersklasse_Start"] = pd.to_numeric(
        hazard_rates["Altersklasse_Start"],
        errors="coerce"
    ).astype("Int64")

    hazard_rates["Altersklasse_Ende"] = pd.to_numeric(
        hazard_rates["Altersklasse_Ende"],
        errors="coerce"
    ).astype("Int64")

    hazard_rates = hazard_rates.dropna(
        subset=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_hazard_rate"
        ]
    )

    # Abwrackprämie: Hazard-Raten für Ziel-Antriebe erhöhen
    hazard_rates["scrappage_target"] = hazard_rates["Antriebsart"].apply(
        lambda x: contains_any_pattern(
            x,
            params["SCRAPPAGE_TARGET_PATTERNS"]
        )
    )

    if scenario == "abwrackpraemie":
        hazard_rates["pooled_hazard_rate"] = np.where(
            hazard_rates["scrappage_target"],
            hazard_rates["pooled_hazard_rate"]
            * params["SCRAPPAGE_HAZARD_MULTIPLIER"],
            hazard_rates["pooled_hazard_rate"]
        )

        hazard_rates["pooled_hazard_rate"] = (
            hazard_rates["pooled_hazard_rate"]
            .clip(lower=0, upper=max_hazard_rate)
        )

    # --------------------------------------------------------
    # 2. Reentry-Quoten vorbereiten
    # --------------------------------------------------------

    reentry_rates = reentry_pooled_df.copy()

    reentry_rates["pooled_reentry_quote_exits"] = pd.to_numeric(
        reentry_rates["pooled_reentry_quote_exits"],
        errors="coerce"
    ).fillna(0)

    reentry_rates["pooled_reentry_quote_exits"] = (
        reentry_rates["pooled_reentry_quote_exits"]
        .clip(lower=0, upper=max_reentry_rate)
    )

    reentry_rates = reentry_rates[
        [
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_reentry_quote_exits"
        ]
    ].copy()

    reentry_rates["Altersklasse_Start"] = pd.to_numeric(
        reentry_rates["Altersklasse_Start"],
        errors="coerce"
    ).astype("Int64")

    reentry_rates["Altersklasse_Ende"] = pd.to_numeric(
        reentry_rates["Altersklasse_Ende"],
        errors="coerce"
    ).astype("Int64")

    reentry_rates = reentry_rates.dropna(
        subset=[
            "Segment",
            "Antriebsart",
            "Altersklasse_Start",
            "Altersklasse_Ende",
            "pooled_reentry_quote_exits"
        ]
    )

    # --------------------------------------------------------
    # 3. Basisverteilung der Neuzulassungen aus Startjahr
    # --------------------------------------------------------

    nzl_work = nzl_df.copy()

    nzl_relevante_spalten = [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Anzahl"
    ]

    nzl_work = nzl_work[nzl_relevante_spalten].copy()

    nzl_work["Berichtsjahr"] = bereinige_berichtsjahr(
        nzl_work,
        "Berichtsjahr",
        bestand_logik=False
    )

    nzl_work["Anzahl"] = pd.to_numeric(
        nzl_work["Anzahl"],
        errors="coerce"
    )

    nzl_work = nzl_work.dropna(subset=nzl_relevante_spalten)

    nzl_start = nzl_work[
        nzl_work["Berichtsjahr"] == start_year
    ].copy()

    nzl_base_distribution = (
        nzl_start
        .groupby(["Segment", "Antriebsart"], as_index=False)["Anzahl"]
        .sum()
        .rename(columns={"Anzahl": "nzl_startjahr"})
    )

    summe_nzl_start = safe_sum(nzl_base_distribution["nzl_startjahr"])

    if summe_nzl_start <= 0:
        raise ValueError(
            f"Für das Startjahr {start_year} wurden keine gültigen "
            "Neuzulassungen gefunden. Die Verteilung neuer Fahrzeuge "
            "kann daher nicht berechnet werden."
        )

    nzl_base_distribution["anteil_original"] = (
        nzl_base_distribution["nzl_startjahr"] / summe_nzl_start
    )

    nzl_base_distribution["ist_verbrenner"] = (
        nzl_base_distribution["Antriebsart"].apply(
            lambda x: ist_verbrenner(x, params)
        )
    )

    nzl_base_distribution["ist_diesel"] = (
        nzl_base_distribution["Antriebsart"].apply(
            lambda x: ist_diesel(x, params)
        )
    )

    nzl_base_distribution = nzl_base_distribution[
        [
            "Segment",
            "Antriebsart",
            "ist_verbrenner",
            "ist_diesel",
            "anteil_original"
        ]
    ].copy()

    # --------------------------------------------------------
    # 4. Ziel-Gesamtbestände vorbereiten
    # --------------------------------------------------------

    fleet_total_work = fleet_total_df.copy()

    fleet_total_work["Jahr"] = pd.to_numeric(
        fleet_total_work["Jahr"],
        errors="coerce"
    ).astype("Int64")

    fleet_total_work["gesamtbestand_target"] = pd.to_numeric(
        fleet_total_work["gesamtbestand_target"],
        errors="coerce"
    )

    fleet_total_work = fleet_total_work.dropna(
        subset=["Jahr", "gesamtbestand_target"]
    )

    target_dict = dict(
        zip(
            fleet_total_work["Jahr"].astype(int),
            fleet_total_work["gesamtbestand_target"]
        )
    )

    fehlende_zieljahre = [
        jahr for jahr in range(start_year, end_year + 1)
        if jahr not in target_dict
    ]

    if fehlende_zieljahre:
        raise ValueError(
            "Für folgende Jahre fehlen Zielbestände in fleet_total_df: "
            + ", ".join(map(str, fehlende_zieljahre))
        )

    # --------------------------------------------------------
    # 5. Startbestand vorbereiten
    # --------------------------------------------------------

    bestand_work = bestand_df.copy()

    bestand_relevante_spalten = [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Jahr der Erstzulassung",
        "Anzahl"
    ]

    bestand_work = bestand_work[bestand_relevante_spalten].copy()

    bestand_work["Berichtsjahr"] = bereinige_berichtsjahr(
        bestand_work,
        "Berichtsjahr",
        bestand_logik=not bestand_berichtsjahr_ist_bereits_angepasst
    )

    bestand_work["Jahr der Erstzulassung"] = pd.to_numeric(
        bestand_work["Jahr der Erstzulassung"],
        errors="coerce"
    ).astype("Int64")

    bestand_work["Anzahl"] = pd.to_numeric(
        bestand_work["Anzahl"],
        errors="coerce"
    )

    bestand_work = bestand_work.dropna(subset=bestand_relevante_spalten)

    current_stock = bestand_work[
        bestand_work["Berichtsjahr"] == start_year
    ].copy()

    if current_stock.empty:
        raise ValueError(
            f"Für das Startjahr {start_year} wurde kein Bestand gefunden. "
            "Prüfe die Spalte Berichtsjahr und die Einstellung "
            "bestand_berichtsjahr_ist_bereits_angepasst."
        )

    current_stock = (
        current_stock
        .groupby(
            ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
            as_index=False
        )["Anzahl"]
        .sum()
        .rename(columns={"Anzahl": "bestand"})
    )

    current_stock["Jahr der Erstzulassung"] = pd.to_numeric(
        current_stock["Jahr der Erstzulassung"],
        errors="coerce"
    ).astype("Int64")

    current_stock["bestand"] = pd.to_numeric(
        current_stock["bestand"],
        errors="coerce"
    ).fillna(0)

    current_stock = current_stock.dropna(
        subset=["Segment", "Antriebsart", "Jahr der Erstzulassung"]
    )

    current_stock = current_stock[
        current_stock["bestand"] > 0
    ].copy()

    # --------------------------------------------------------
    # 6. Simulation Jahr für Jahr
    # --------------------------------------------------------

    scenario_results_by_year = {}
    summary_rows = []

    scenario_results_by_year[start_year] = {
        "abgaenge": pd.DataFrame(),
        "rueckkehrer": pd.DataFrame(),
        "fehlende_fahrzeuge": pd.DataFrame({
            "Berichtsjahr": [start_year],
            "fehlende_fahrzeuge": [
                max(target_dict[start_year] - safe_sum(current_stock["bestand"]), 0)
            ]
        }),
        "neue_fahrzeuge": pd.DataFrame(),
        "finaler_bestand": current_stock.assign(Berichtsjahr=start_year).rename(
            columns={"bestand": "finaler_bestand"}
        )[
            [
                "Berichtsjahr",
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "finaler_bestand"
            ]
        ].copy()
    }

    summary_rows.append({
        "Szenario": scenario,
        "Jahr": start_year,
        "zielbestand": target_dict[start_year],
        "bestand_vor_neuen_fahrzeugen": safe_sum(current_stock["bestand"]),
        "fehlende_fahrzeuge": max(
            target_dict[start_year] - safe_sum(current_stock["bestand"]),
            0
        ),
        "finaler_bestand": safe_sum(current_stock["bestand"])
    })

    for jahr in range(start_year, end_year):

        folgejahr = jahr + 1

        # ----------------------------------------------------
        # 6.1 Alter und Altersklasse berechnen
        # ----------------------------------------------------

        projection_base = current_stock.copy()

        projection_base = berechne_altersklasse(
            projection_base,
            aktuelles_jahr=jahr,
            breite=bin_width
        )

        # ----------------------------------------------------
        # 6.2 Hazard- und Reentry-Raten ergänzen
        # ----------------------------------------------------

        projection_base = projection_base.merge(
            hazard_rates,
            on=[
                "Segment",
                "Antriebsart",
                "Altersklasse_Start",
                "Altersklasse_Ende"
            ],
            how="left"
        )

        projection_base = projection_base.merge(
            reentry_rates,
            on=[
                "Segment",
                "Antriebsart",
                "Altersklasse_Start",
                "Altersklasse_Ende"
            ],
            how="left"
        )

        projection_base["pooled_hazard_rate"] = (
            projection_base["pooled_hazard_rate"]
            .fillna(0)
            .clip(lower=0, upper=max_hazard_rate)
        )

        projection_base["pooled_reentry_quote_exits"] = (
            projection_base["pooled_reentry_quote_exits"]
            .fillna(0)
            .clip(lower=0, upper=max_reentry_rate)
        )

        # ----------------------------------------------------
        # 6.3 Exits, Reentries und Bestand vor neuen Fahrzeugen
        # ----------------------------------------------------

        projection_base["exits"] = (
            projection_base["bestand"]
            * projection_base["pooled_hazard_rate"]
        )

        projection_base["reentries"] = (
            projection_base["exits"]
            * projection_base["pooled_reentry_quote_exits"]
        )

        projection_base["bestand_vor_neuen_fahrzeugen"] = (
            projection_base["bestand"]
            - projection_base["exits"]
            + projection_base["reentries"]
        ).clip(lower=0)

        # ----------------------------------------------------
        # 6.4 Diesel-Fahrverbot:
        #     ab Ban-Year Diesel-Bestand auf 0 setzen
        # ----------------------------------------------------

        if scenario == "diesel_fahrverbot" and folgejahr >= params["DIESEL_BAN_YEAR"]:
            projection_base["ist_diesel"] = projection_base["Antriebsart"].apply(
                lambda x: ist_diesel(x, params)
            )

            projection_base["bestand_vor_neuen_fahrzeugen"] = np.where(
                projection_base["ist_diesel"],
                0,
                projection_base["bestand_vor_neuen_fahrzeugen"]
            )

        bestehender_bestand = safe_sum(
            projection_base["bestand_vor_neuen_fahrzeugen"]
        )

        zielbestand = target_dict[folgejahr]

        fehlende_fahrzeuge = max(
            zielbestand - bestehender_bestand,
            0
        )

        # ----------------------------------------------------
        # 6.5 Szenariospezifische Neuzulassungsverteilung anwenden
        # ----------------------------------------------------

        nzl_distribution_scenario = passe_nzl_verteilung_an(
            base_distribution=nzl_base_distribution,
            jahr=folgejahr,
            scenario=scenario,
            params=params
        )

        new_vehicles = nzl_distribution_scenario.copy()

        new_vehicles["Jahr der Erstzulassung"] = folgejahr
        new_vehicles["neue_fahrzeuge"] = (
            fehlende_fahrzeuge * new_vehicles["anteil_szenario"]
        )

        new_vehicles = new_vehicles[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "ist_verbrenner",
                "ist_diesel",
                "anteil_original",
                "anteil_szenario",
                "neue_fahrzeuge"
            ]
        ].copy()

        # ----------------------------------------------------
        # 6.6 Finalen Bestand des Folgejahres erstellen
        # ----------------------------------------------------

        stock_existing_next = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "bestand_vor_neuen_fahrzeugen"
            ]
        ].copy()

        stock_existing_next = stock_existing_next.rename(
            columns={"bestand_vor_neuen_fahrzeugen": "bestand"}
        )

        stock_new_next = new_vehicles[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "neue_fahrzeuge"
            ]
        ].copy()

        stock_new_next = stock_new_next.rename(
            columns={"neue_fahrzeuge": "bestand"}
        )

        current_stock_next = pd.concat(
            [
                stock_existing_next,
                stock_new_next
            ],
            ignore_index=True
        )

        current_stock_next["bestand"] = pd.to_numeric(
            current_stock_next["bestand"],
            errors="coerce"
        ).fillna(0)

        current_stock_next = (
            current_stock_next
            .groupby(
                ["Segment", "Antriebsart", "Jahr der Erstzulassung"],
                as_index=False
            )["bestand"]
            .sum()
        )

        current_stock_next = current_stock_next[
            current_stock_next["bestand"] > 0
        ].copy()

        finaler_bestand = safe_sum(current_stock_next["bestand"])

        # ----------------------------------------------------
        # 6.7 Ergebnisse speichern
        # ----------------------------------------------------

        abgaenge_df = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "exits"
            ]
        ].copy()

        abgaenge_df["Berichtsjahr"] = folgejahr

        rueckkehrer_df = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "reentries"
            ]
        ].copy()

        rueckkehrer_df["Berichtsjahr"] = folgejahr

        bestand_vor_neu_df = projection_base[
            [
                "Segment",
                "Antriebsart",
                "Jahr der Erstzulassung",
                "bestand_vor_neuen_fahrzeugen"
            ]
        ].copy()

        bestand_vor_neu_df["Berichtsjahr"] = folgejahr

        fehlende_fahrzeuge_df = pd.DataFrame({
            "Berichtsjahr": [folgejahr],
            "fehlende_fahrzeuge": [fehlende_fahrzeuge]
        })

        finaler_bestand_df = current_stock_next.copy()
        finaler_bestand_df["Berichtsjahr"] = folgejahr
        finaler_bestand_df = finaler_bestand_df.rename(
            columns={"bestand": "finaler_bestand"}
        )

        scenario_results_by_year[folgejahr] = {
            "abgaenge": abgaenge_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "exits"
                ]
            ].copy(),

            "rueckkehrer": rueckkehrer_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "reentries"
                ]
            ].copy(),

            "bestand_vor_neuen_fahrzeugen": bestand_vor_neu_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "bestand_vor_neuen_fahrzeugen"
                ]
            ].copy(),

            "fehlende_fahrzeuge": fehlende_fahrzeuge_df.copy(),

            "neue_fahrzeuge": new_vehicles.assign(
                Berichtsjahr=folgejahr
            )[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "ist_verbrenner",
                    "ist_diesel",
                    "anteil_original",
                    "anteil_szenario",
                    "neue_fahrzeuge"
                ]
            ].copy(),

            "finaler_bestand": finaler_bestand_df[
                [
                    "Berichtsjahr",
                    "Segment",
                    "Antriebsart",
                    "Jahr der Erstzulassung",
                    "finaler_bestand"
                ]
            ].copy()
        }

        summary_rows.append({
            "Szenario": scenario,
            "Jahr": folgejahr,
            "zielbestand": zielbestand,
            "bestand_vor_neuen_fahrzeugen": bestehender_bestand,
            "fehlende_fahrzeuge": fehlende_fahrzeuge,
            "finaler_bestand": finaler_bestand
        })

        # Finaler Bestand des Folgejahres wird Startbestand für nächste Iteration
        current_stock = current_stock_next.copy()

    # --------------------------------------------------------
    # 7. Summary-Tabelle
    # --------------------------------------------------------

    scenario_summary_df = pd.DataFrame(summary_rows)

    scenario_summary_df["abweichung_final_vs_ziel"] = (
        scenario_summary_df["finaler_bestand"]
        - scenario_summary_df["zielbestand"]
    )

    scenario_summary_df = scenario_summary_df[
        [
            "Szenario",
            "Jahr",
            "zielbestand",
            "bestand_vor_neuen_fahrzeugen",
            "fehlende_fahrzeuge",
            "finaler_bestand",
            "abweichung_final_vs_ziel"
        ]
    ].copy()

    return scenario_results_by_year, scenario_summary_df


# ------------------------------------------------------------
# Alle Szenarien automatisch durchrechnen
# ------------------------------------------------------------

policy_scenario_outputs = {}

for scenario in PARAMS_EXT["SCENARIOS"]:

    scenario_results_by_year, scenario_summary_df = run_policy_scenario(
        scenario=scenario,
        params=PARAMS_EXT,
        bestand_df=bestand_df,
        nzl_df=nzl_df,
        hazard_pooled_df=hazard_pooled_df,
        reentry_pooled_df=reentry_pooled_df,
        fleet_total_df=fleet_total_df,
        bestand_berichtsjahr_ist_bereits_angepasst=True
    )

    policy_scenario_outputs[scenario] = {
        "results_by_year": scenario_results_by_year,
        "summary": scenario_summary_df
    }


# ------------------------------------------------------------
# Kontrolle ausgeben
# ------------------------------------------------------------

print("Berechnete Szenarien:")
print(list(policy_scenario_outputs.keys()))

print("\nSummary baseline:")
display(policy_scenario_outputs["baseline"]["summary"].head())

print("\nSummary diesel_fahrverbot:")
display(policy_scenario_outputs["diesel_fahrverbot"]["summary"].head())

print("\nSummary abwrackpraemie:")
display(policy_scenario_outputs["abwrackpraemie"]["summary"].head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import math

# ------------------------------------------------------------
# 1. Szenario-Output auswählen
# ------------------------------------------------------------
# Falls du die erweiterten Policy-Szenarien berechnet hast:
scenario_data = policy_scenario_outputs

# Falls du stattdessen die vorherigen zwei Szenarien verwenden möchtest:
# scenario_data = scenario_outputs


# ------------------------------------------------------------
# 2. Hilfsfunktion: Pivot-Tabelle je Szenario erstellen
# ------------------------------------------------------------

def create_stock_by_drive_pivot(scenario_data, scenario_name):
    """
    Erstellt für ein Szenario eine Pivot-Tabelle:
    Zeilen: Jahre
    Spalten: Antriebsarten
    Werte: Fahrzeugbestand
    """

    stock_by_drive_rows = []

    results_by_year = scenario_data[scenario_name]["results_by_year"]

    for jahr, year_results in results_by_year.items():

        stock_df = year_results["finaler_bestand"].copy()

        # Bestandsspalte robust erkennen
        if "finaler_bestand" in stock_df.columns:
            bestand_spalte = "finaler_bestand"
        elif "bestand" in stock_df.columns:
            bestand_spalte = "bestand"
        else:
            raise ValueError(
                f"Im Szenario '{scenario_name}' wurde für das Jahr {jahr} "
                "keine Bestandsspalte gefunden. Erwartet wurde "
                "'finaler_bestand' oder 'bestand'."
            )

        stock_df[bestand_spalte] = pd.to_numeric(
            stock_df[bestand_spalte],
            errors="coerce"
        ).fillna(0)

        stock_by_drive = (
            stock_df
            .groupby("Antriebsart", as_index=False)[bestand_spalte]
            .sum()
            .rename(columns={bestand_spalte: "fahrzeugbestand"})
        )

        stock_by_drive["Jahr"] = jahr

        stock_by_drive_rows.append(stock_by_drive)

    if len(stock_by_drive_rows) == 0:
        raise ValueError(f"Für Szenario '{scenario_name}' wurden keine Bestandsdaten gefunden.")

    stock_by_drive_year_df = pd.concat(
        stock_by_drive_rows,
        ignore_index=True
    )

    stock_by_drive_pivot = (
        stock_by_drive_year_df
        .pivot_table(
            index="Jahr",
            columns="Antriebsart",
            values="fahrzeugbestand",
            aggfunc="sum",
            fill_value=0
        )
        .sort_index()
    )

    return stock_by_drive_pivot


# ------------------------------------------------------------
# 3. Pivot-Tabellen für alle Szenarien erstellen
# ------------------------------------------------------------

scenario_names = list(scenario_data.keys())

if len(scenario_names) == 0:
    raise ValueError("Es wurden keine Szenarien in scenario_data gefunden.")

scenario_pivots = {}

for scenario_name in scenario_names:
    scenario_pivots[scenario_name] = create_stock_by_drive_pivot(
        scenario_data,
        scenario_name
    )


# ------------------------------------------------------------
# 4. Spalten aller Pivot-Tabellen angleichen
# ------------------------------------------------------------

all_drive_types = sorted(
    set().union(*[set(pivot.columns) for pivot in scenario_pivots.values()])
)

for scenario_name in scenario_names:
    scenario_pivots[scenario_name] = scenario_pivots[scenario_name].reindex(
        columns=all_drive_types,
        fill_value=0
    )


# ------------------------------------------------------------
# 5. Gemeinsame y-Achse bestimmen
# ------------------------------------------------------------

y_max = max(
    pivot.sum(axis=1).max()
    for pivot in scenario_pivots.values()
)

y_max = y_max * 1.05


# ------------------------------------------------------------
# 6. Raster für Diagramme erstellen
# ------------------------------------------------------------

n_scenarios = len(scenario_names)

ncols = 1 if n_scenarios == 1 else 2
nrows = math.ceil(n_scenarios / ncols)

fig, axes = plt.subplots(
    nrows=nrows,
    ncols=ncols,
    figsize=(9 * ncols, 6 * nrows),
    sharey=True
)

# axes immer als flache Liste behandeln
if n_scenarios == 1:
    axes = [axes]
else:
    axes = axes.flatten()


# ------------------------------------------------------------
# 7. Gestapelte Balkendiagramme je Szenario plotten
# ------------------------------------------------------------

for ax, scenario_name in zip(axes, scenario_names):

    scenario_pivots[scenario_name].plot(
        kind="bar",
        stacked=True,
        ax=ax,
        legend=False
    )

    ax.set_title(scenario_name)
    ax.set_xlabel("Year")
    ax.set_ylabel("Vehicle stock")
    ax.set_ylim(0, y_max)
    ax.tick_params(axis="x", rotation=45)


# Leere Diagrammfelder ausblenden
for ax in axes[n_scenarios:]:
    ax.set_visible(False)


# ------------------------------------------------------------
# 8. Gemeinsame Legende rechts außerhalb platzieren
# ------------------------------------------------------------

handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    title="Drive type",
    bbox_to_anchor=(1.02, 0.5),
    loc="center left"
)

plt.tight_layout(rect=[0, 0, 0.85, 1])
plt.show()


# ------------------------------------------------------------
# 9. Kontrolle: erste Zeilen der Pivot-Tabellen anzeigen
# ------------------------------------------------------------

for scenario_name in scenario_names:
    print(f"Pivot-Tabelle {scenario_name}:")
    display(scenario_pivots[scenario_name].head())
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Fleet-Size-Szenarien
# ------------------------------------------------------------

FLEET_CHANGE_SCENARIOS = {
    "fleet_plus_5": 5.0,
    "fleet_plus_10": 10.0,
    "fleet_plus_15": 15.0,
}

# Policy-Modus, der mit den Fleet-Size-Pfaden kombiniert werden soll
# Mögliche Werte z.B.: "baseline", "verbrenneraus", "diesel_fahrverbot", "abwrackpraemie"
FLEET_SCENARIO_MODE = "baseline"

# Art der Fortschreibung des Zielbestandspfads:
# "linear" = gleicher absoluter Zuwachs pro Jahr
# "growth" = gleiche jährliche prozentuale Wachstumsrate
FLEET_TARGET_METHOD = "linear"


# ------------------------------------------------------------
# Hilfsfunktionen
# ------------------------------------------------------------

def bereinige_berichtsjahr_fleet(df, spalte="Berichtsjahr", bestand_logik=False):
    """
    Wandelt Berichtsjahr robust in ein Jahr um.
    Funktioniert bei Jahreszahlen wie 2025 und bei Datumswerten wie 01.01.2026.

    Wenn bestand_logik=True:
    Ein Bestand mit Datum 01.01.2026 wird als Bestand des Jahres 2025 interpretiert.
    Dann wird minus 1 gerechnet.
    """
    s = df[spalte]

    jahr_num = pd.to_numeric(s, errors="coerce")
    jahr_date = pd.to_datetime(s, errors="coerce", dayfirst=True).dt.year

    jahr = jahr_num.fillna(jahr_date)

    if bestand_logik:
        jahr = jahr - 1

    return jahr.astype("Int64")


def get_startbestand_from_bestand_df(
    bestand_df,
    start_year=2025,
    bestand_berichtsjahr_ist_bereits_angepasst=True
):
    """
    Berechnet den Gesamtbestand im Startjahr aus bestand_df.
    """
    bestand_work = bestand_df.copy()

    relevante_spalten = [
        "Berichtsjahr",
        "Segment",
        "Antriebsart",
        "Jahr der Erstzulassung",
        "Anzahl"
    ]

    bestand_work = bestand_work[relevante_spalten].copy()

    bestand_work["Berichtsjahr"] = bereinige_berichtsjahr_fleet(
        bestand_work,
        "Berichtsjahr",
        bestand_logik=not bestand_berichtsjahr_ist_bereits_angepasst
    )

    bestand_work["Anzahl"] = pd.to_numeric(
        bestand_work["Anzahl"],
        errors="coerce"
    )

    bestand_work = bestand_work.dropna(subset=["Berichtsjahr", "Anzahl"])

    startbestand = bestand_work.loc[
        bestand_work["Berichtsjahr"] == start_year,
        "Anzahl"
    ].sum()

    if pd.isna(startbestand) or startbestand <= 0:
        raise ValueError(
            f"Für das Startjahr {start_year} konnte kein gültiger Startbestand "
            "aus bestand_df berechnet werden. Prüfe die Spalte Berichtsjahr "
            "und die Einstellung bestand_berichtsjahr_ist_bereits_angepasst."
        )

    return startbestand


def create_fleet_target_path(
    bestand_df,
    percent_change_2050,
    start_year=2025,
    end_year=2050,
    method="linear",
    bestand_berichtsjahr_ist_bereits_angepasst=True
):
    """
    Erzeugt einen Zielbestandspfad vom Startjahr bis zum Zieljahr.

    percent_change_2050:
    - 5.0 bedeutet +5 % bis 2050 gegenüber dem Startbestand.
    - 10.0 bedeutet +10 % bis 2050 gegenüber dem Startbestand.

    method:
    - "linear": gleiche absolute Veränderung pro Jahr
    - "growth": gleiche jährliche Wachstumsrate

    Rückgabe:
    DataFrame mit Spalten:
    - Jahr
    - gesamtbestand_target
    - Gesamtbestand_target
    """

    startbestand = get_startbestand_from_bestand_df(
        bestand_df=bestand_df,
        start_year=start_year,
        bestand_berichtsjahr_ist_bereits_angepasst=bestand_berichtsjahr_ist_bereits_angepasst
    )

    zielbestand = startbestand * (1 + percent_change_2050 / 100)

    jahre = np.arange(start_year, end_year + 1)
    n_periods = end_year - start_year

    if method == "linear":
        target_values = np.linspace(
            startbestand,
            zielbestand,
            len(jahre)
        )

    elif method == "growth":
        if startbestand <= 0:
            raise ValueError(
                "Für die Wachstumsraten-Methode muss der Startbestand größer als 0 sein."
            )

        annual_growth_rate = (zielbestand / startbestand) ** (1 / n_periods) - 1

        target_values = startbestand * (
            (1 + annual_growth_rate) ** np.arange(len(jahre))
        )

    else:
        raise ValueError(
            "method muss entweder 'linear' oder 'growth' sein."
        )

    fleet_target_path_df = pd.DataFrame({
        "Jahr": jahre,
        "gesamtbestand_target": target_values
    })

    # Zusätzliche Spalte gemäß deiner gewünschten Benennung.
    # run_policy_scenario nutzt weiterhin "gesamtbestand_target".
    fleet_target_path_df["Gesamtbestand_target"] = (
        fleet_target_path_df["gesamtbestand_target"]
    )

    return fleet_target_path_df


# ------------------------------------------------------------
# Ursprünglichen fleet_total_df sichern
# ------------------------------------------------------------

fleet_total_df_original = fleet_total_df.copy()


# ------------------------------------------------------------
# Fleet-Size-Szenarien berechnen
# ------------------------------------------------------------

fleet_size_scenario_outputs = {}
fleet_size_target_paths = {}

try:
    for fleet_scenario_name, percent_change in FLEET_CHANGE_SCENARIOS.items():

        # Temporärer Zielbestandspfad für dieses Fleet-Size-Szenario
        temporary_fleet_total_df = create_fleet_target_path(
            bestand_df=bestand_df,
            percent_change_2050=percent_change,
            start_year=PARAMS_EXT["START_YEAR"],
            end_year=PARAMS_EXT["END_YEAR_MULTI"],
            method=FLEET_TARGET_METHOD,
            bestand_berichtsjahr_ist_bereits_angepasst=True
        )

        # Zusätzlich separat speichern
        fleet_size_target_paths[fleet_scenario_name] = (
            temporary_fleet_total_df.copy()
        )

        # Szenario mit temporärem Zielbestandspfad berechnen
        scenario_results_by_year, scenario_summary_df = run_policy_scenario(
            scenario=FLEET_SCENARIO_MODE,
            params=PARAMS_EXT,
            bestand_df=bestand_df,
            nzl_df=nzl_df,
            hazard_pooled_df=hazard_pooled_df,
            reentry_pooled_df=reentry_pooled_df,
            fleet_total_df=temporary_fleet_total_df,
            bestand_berichtsjahr_ist_bereits_angepasst=True
        )

        fleet_size_scenario_outputs[fleet_scenario_name] = {
            "percent_change_2050": percent_change,
            "policy_mode": FLEET_SCENARIO_MODE,
            "target_path": temporary_fleet_total_df.copy(),
            "results_by_year": scenario_results_by_year,
            "summary": scenario_summary_df
        }

finally:
    # Ursprünglichen fleet_total_df wiederherstellen
    fleet_total_df = fleet_total_df_original.copy()


# ------------------------------------------------------------
# Kontrolle ausgeben
# ------------------------------------------------------------

print("Berechnete Fleet-Size-Szenarien:")

for scenario_name, output in fleet_size_scenario_outputs.items():
    print(
        f"- {scenario_name}: "
        f"{output['percent_change_2050']} % bis 2050, "
        f"Policy-Modus: {output['policy_mode']}"
    )

print("\nSummary fleet_plus_5:")
display(fleet_size_scenario_outputs["fleet_plus_5"]["summary"].head())

print("\nSummary fleet_plus_10:")
display(fleet_size_scenario_outputs["fleet_plus_10"]["summary"].head())

print("\nSummary fleet_plus_15:")
display(fleet_size_scenario_outputs["fleet_plus_15"]["summary"].head())
```
